### &#9745; Exercise 1

`17/8` does a 'normal' division, as you can see by executing the following cell:

```python
17/8
```

In contrast, `17//8` only computes the integer part of the ratio $\frac{17}{8}$:

```python
17//8
```

A number can be raised to some power with the double star `**` operator. This allows us to compute $7^3+2^3$ as follows:

```python
7**3+2**3
```

Numbers are multiplied with a single star `*`:

```python
1*2*3*4*5*6*7*8
```

To compute $2^{3+4^2}$ we again use `**`, but care is required to bracket this expression correctly:

```python
2**(3+4**2)
```

If we leave out the brackets, we get a completely different result.

```python
2**3+4**2
```

To compute $10000^4-1$ we can use:

```python
10000**4-1
```

### &#9745; Exercise 3

The following code creates a Python list with the numbers $1,3,5,7,9$ and then changes the pen-ultimate element to -11:

```python
some_list = [1,3,5,7,9]
print(some_list)
some_list[-2] = -11
print(some_list)
```

Instead of accessing the pen-ultimate element as `some_list[-2]` we could also have used `some_list[3]`, but the former is more elegant since we are couting directly from the back, i.e. using the index `-2` gives the correct result irrespective of the length of the list.

To compute $e^{5}$, $\sin(\pi/7)$ and $\sqrt{e^{3}+2}$ we can use the numpy functions `exp`, `sin` and `sqrt` and the numpy constant `pi`. Of course this only works if we first import the numpy library

```python
import numpy as np
```

```python
np.exp(5)
```

```python
np.sin(np.pi/7)
```

```python
np.sqrt(np.exp(3)+2)
```

As the following code demonstates, when applied to a list the `np.log2()` function takes the logarithm of each element:

```python
np.log2([1,2,4,8,32,512])
```

The same applies to many other numpy functions. If `np.f()` is a numpy function which computes $f(x)$ for a given number $x$ and $a=[a_0,a_1,\dots,a_{n-1}]$ is a list, then `np.f(a)` returns the list $[f(a_0),f(a_1),\dots,f(a_{n-1})]$. In other words, the function is applied elementwise.

### &#9745; Exercise 5

The matrices $\begin{pmatrix}1&2\\3&4\end{pmatrix}$ and $D=\begin{pmatrix}1&2\\3&4\\5&6\end{pmatrix}$ are created as follows:

```python
A = np.array([[1,2],[3,4]])
D = np.array([[1,2],[3,4],[5,6]])
```

The matrix product $A\circ A$ is computed with the `@` operator:

```python
A@A
```

In contrast, the `*` operator computes the elementwise product:

```python
A*A
```

$A\circ D$ is not valid since this is the matrix-product of a $2\times 2$ matrix and $3\times 2$ matrix:

```python
A@D
```

$D\circ A$ is valid since this is the matrix-product of a $3\times 2$ matrix and $2\times 2$ matrix. The result is another $3\times 2$ matrix:

```python
D@A
```

The elementwise products $A\odot D$ and $D\odot A$ are both invalid, since the shapes of the matrices do not agree:

```python
A*D
```

```python
D*A
```

First, create the matrix $B = \begin{pmatrix}1&8\\2&9\\-1&4\\0&2\end{pmatrix}$ and the vector $v = \begin{pmatrix}-3\\4\\6\\-2\end{pmatrix}$


```python
B = np.array([[1,8],[2,9],[-1,4],[0,2]])
v = np.array([-3,4,6,-2])
```

The product $B^Tv$ can now be computed as

```python
B.transpose()@v
```

```python
np.matmul(B.transpose(),v)
```

```python
B.transpose().dot(v)
```

Since $B^Tv=(v^TB)^T$ there is another way of doing this as well:

```python
(v.transpose()@B).transpose()
```

### What does "=" actually mean?
It is important to note that in Python a single equals sign (`=`) has a different meaning than in a mathematical statement. In Python `a=5` means: *''Assign the name "a" to the number 5''*. Therefore you can write things like
```Python
a = 3
a = a + 2
```
which means: *''First assign the name "a" to the number 3. Then assign the name "a" to the number a+2, which is just 3+2.''* Note that mathematically the statement $a=a+2$ has a completely different meaning: if $a$ is an integer number, then mathematically this is a false statement since $a+2$ can never be the same as $a$. For this reason, we will often write "$a\mapsto a+2$" instead when we interpret the Python statement `a=a+2` mathematically in so-called *pseudocode*. The pseudocode for the two lines above would be:

$$
\begin{aligned}
a &\mapsto 3\\
a &\mapsto a + 2
\end{aligned}
$$

### Comparing two numbers with "=="

### Manipulating matrices and vectors with numpy
In this section we will have a look at working with vectors and matrices in Python. Before reading on, you should make sure that you are familiar with the content of the [notebook on vectors and matrices](./Lecture_01a_VectorsMatrices.ipynb), which contains the necessary mathematical background.

Mathematically, the entries of a list could be the components of a vector. For example, the vectors
$$
\boldsymbol{v} = \begin{pmatrix}6\\-2\\4\end{pmatrix}\qquad\text{and}\qquad \boldsymbol{w} = \begin{pmatrix}1\\0\\9\end{pmatrix}
$$
could be represented by the Python lists `v=[6,-2,4]` and `w=[1,0,9]`. However, to do something useful with those vector, for example add them or multiply them by a matrix, we first need to tell Python that those really are vectors. For this, we use the `numpy` function `np.array()`, which converts a list to a vector. Once we have done this, we can add $v$ and $w$ to get the expected result:

````python
v = np.array([6,-2,4])
w = np.array([1,0,9])
print (v)
print (w)
print (v+w)
````

If you get an error message, recall that you need to import the numpy library first, so scroll back up and execute the cell which contains the line

```Python
import numpy as np
```

What happens if we just add the lists, without converting them to vectors first? Well, let's find out:

````python
v = [6,-2,4]
w = [1,0,9]
print (v)
print (w)
print (v+w)
````

So in this case Python has actually *''added''* the lists by concatenating them! This might be useful in some other contexts (so for the moment just remember that Python allows us to concatenate lists with the `+` operator) but not here.

This example shows that we need to be very explicit when we instruct Python to do something: we need to tell the Python interpreter that the numbers $6,-2,4$ and $1,0,9$ represent a vector, which has certain mathematical properties. In computer science, we say that the list object `[6,-2,4]` and the vector object `np.array([6,-2,4])` have different *types*. We will learn more about different types in one of the later lectures.

We can define matrices in a similar way. For example, Let's imagine we want to construct a $2\times 3$ matrix

$$
A = \begin{pmatrix} 1 & 0 & 8\\-2 & 1 & 3\end{pmatrix}

To do this, we call `np.array()` with a list of two lists, which contain the two lines of the matrix $A$:

````python
A = np.array([[1,0,8],[-2,1,3]])
print(A)
````

We can multiply a matrix and a vector with the `@` symbol. To compute the vector $\boldsymbol{t}=A\boldsymbol{v}$ use:

````python
t = A@v
print(t)
````

C = \begin{pmatrix} 1 & 0 & 8 \\ -2 & 1 & 3 \end{pmatrix}

### Matrix- and componentwise multiplication
Recall that more generally, the product of a $n\times m$ matrix $A$ and an $m\times r$ matrix $B$ is an $n\times r$ matrix $C=A\circ B$, such that the entry $C_{ij}$ is given as

$$
C_{ij} = \sum_{k=0}^{m-1} A_{ik}B_{kj}\qquad\text{for $i=0,1,2,\dots,n-1$ and $j=0,1,2,\dots,r-1$}
$$

In Python we always start counting from $0$ to $n-1$ instead of $1$ to $n$. Often we leave out the dot, and just write $C=AB$.

Matrix multiplication is not the same as componentwise multiplication, which is defined as follows: Let $D$ and $F$ be $n\times m$ matrices. Then the componentwise product (denoted by $\odot$) is $G=D\odot F$ such that

$$
G_{ij} = D_{ij}F_{ij}\qquad\text{for $i=0,1,2,\dots,n-1$ and $j=0,1,2,\dots,r-1$}
$$

For example, for the matrices

$$
D = \begin{pmatrix}1 & 5 \\ 2 & -4\end{pmatrix}\quad\text{and}\quad
F = \begin{pmatrix}-2 & 1 \\ 3 & 0\end{pmatrix}
$$

the matrix-product is

$$
DF = \begin{pmatrix} 1\cdot (-2) + 5\cdot 3 & 1\cdot 1 + 5 \cdot 0 \\ 2\cdot(-2) + (-4)\cdot 3 & 2\cdot 1 + (-4)\cdot 0\end{pmatrix} = \begin{pmatrix}13 & 1 \\ -16 & 2\end{pmatrix}
$$

whereas the componentwise product (denoted by "$\odot$") is

$$
D\odot F = \begin{pmatrix} 1\cdot (-2)  & 5 \cdot 1 \\ 2\cdot 3 & (-4)\cdot 0\end{pmatrix} = \begin{pmatrix}-2 & 5 \\ 6 & 0\end{pmatrix}
$$

In Python, the componentwise product is computed with the `*` operator. Consider the following example:

````python
C = np.asarray([[1,5],[2,-4]])
D = np.asarray([[-2,1],[3,0]])
print(C@D)
print(C*D)
````

This matters if we write down expressions such as `C**3`, which will compute the third power **componentwise**. To raise the matrix $C$ to the third power, i.e. compute $C\cdot C\cdot C$ (instead of $C\circ C\circ C$), we need to use `np.linalg.matrix_power()`. This is illustrated in this example:

````python
print(C**3)
print(C*C*C)
print(C@C@C)
print(np.linalg.matrix_power(C,3))
````

We will look at a clever way of raising a square matrix to some power later in this course.

Mathematically, for a $n\times m$ matrix $A$ the transpose matrix $A^T$ is the $m\times n$ matrix with entries

$$
A^T_{ij} = A_{ji}\qquad\text{for $i=1,2,\dots,m$ and $j=1,2,\dots,n$}
$$

For the $2 \times 3$ matrix $A$ above the $3\times 2$ matrix $A^T$ given by

$$
A^T = \begin{pmatrix} 1 & -2 \\ 0 & 1 \\8 & 3\end{pmatrix}
$$

In Python we can use the numpy function `np.transpose()` to transpose matrices:

````python
np.transpose(A)
````

### A first look at objects
Actually, it turns out there is a different way of doing this. Instead of
```Python
np.transpose(A)
```
we can also write

````python
A.transpose()
````

Why does this work and what is the difference between those two statements? In the first case, the variable `A` assumes a **passive role**: we take the `numpy` function `np.transpose()` and feed it `A`, which holds the matrix entries. In the second case, however, the variable `A` takes on an **active role**, which is signified by the notation `A.do_something()`. This works because `A` is actually a so-called *object* of a certain *type* (in this case it is a matrix object, known as an `ndarray` in `numpy`). In addition to storing some values, `A` "owns" certain functions. Python is a so-called *Object Oriented Language*, and we will see in the second semester why this makes Python so powerful.

Here is another example: to compute the matrix-vector product $A\boldsymbol{v}$, we can use the following, instead of `A@v` or `np.matmul(A,v)`:

````python
A.dot(v)
````

We can think about this as follows: Imagine we are the matrix $A$. In addition to having access to the matrix components $A_{ij}$, we also know how to compute the product with some vector $\boldsymbol{v}$ that we are given. This is conceptually different to `np.matmul(A,v)`, where `numpy` does the computation and $A$ just provides the data.

In addition to owning functions, objects can also have properties. For example, a matrix object stores information on its shape. For a $n\times m$ matrix $A$, calling `A.shape` returns $n$ and $m$:

````python
A.shape
````

Let's look at the shape of the transpose of $A$, which can be done as follows:

````python
A.transpose().shape
````

Here we used the dot twice: First, we construct a new object `A.transpose()` by calling the `transpose()` function, and then we ask for the shape of this.

Note that $n$ and $m$ are returned in round brackets `(2,3)`, and not as a list `[2,3]`. Round-bracket lists are known as "tuples", and they behave similar to lists, but there are subtle differences. We can, for example, access entries of the tuple `(n,m)` with the index notation `[]`:

````python
A.shape[0]
````

Since `A.shape` returns two numbers, it is also possible to write the following:

````python
nrow, ncol = A.shape
print(nrow)
print(ncol)
````

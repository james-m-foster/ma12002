## Summation and multiplication
First, we will introduce some notation that we will use throughout this course to write down sums and products. Let $a_0, a_1, a_2, \dots, a_n$ be some numbers. We use the $\sum$-symbol (''$\Sigma$'' is capital Greek letter Sigma) to denote the their sum:

$$
a_0+a_1+\dots+a_n = \sum_{j=0}^{n} a_j
$$

Here $j$ is called the *summation index*. We can also sum only a subset of those numbers, for example

$$
a_3+a_4+\dots+a_{n-3}+a_{n-2} = \sum_{j=3}^{n-2} a_j
$$

Note that the summation index and the lower bound are written below the $\sum$ symbol whereas the upper bound is written above the $\sum$ symbol. Observe, incidentally, that we can write the same sum as

$$
a_3+a_4+\dots+a_{n-3}+a_{n-2} = \sum_{k=0}^{n-5} a_{k+3}
$$

by chosing a different summation index.

The product is denoted with the $\prod$-symbol (''$\Pi$'' is the capital Greek letter Pi):

$$
a_0\cdot a_1\cdot \dots\cdot  a_n = \prod_{j=0}^{n} a_j
$$

Again, we can only multiply some numbers, e.g.

$$
a_2\cdot a_3 \cdot \cdots \cdot a_{n-2}\cdot a_{n-1} = \prod_{j=2}^{n-1} a_j
$$

## Vectors
For the purpose of this course, an $n$-dimensional vector contains $n$ numbers which are written on top of each other inside round brackets. Consider, for example, the three-dimensional vector $\boldsymbol{u}$ and the four-dimensional vector $\boldsymbol{v}$ (we will always denote vectors with **bold** letters, in other courses they might be written with an arrow $\vec{u}$ or an underline $\underline{u}$):

$$
\boldsymbol{u} = \begin{pmatrix}2\\-4\\1\end{pmatrix}\qquad
\boldsymbol{v} = \begin{pmatrix}3\\8\\9\\-3\end{pmatrix}
$$

The numbers that make up a vector are called the *components* or *entries* of a vector; an $n$-dimensional vector has $n$ components. For the vector $\boldsymbol{u}$ we will label those components as $u_0,u_1,\dots,u_{n-1}$. Note that we start counting at zero and go to $n-1$ - the reason for this will become clear when we work with vectors in Python.

So in general, an $n$-dimensional vector can be written like this:

$$
\boldsymbol{u} = \begin{pmatrix}u_0\\u_1\\u_2\\\vdots\\u_{n-2}\\u_{n-1}\end{pmatrix}\qquad
$$

### Common vector operations
#### Scaling vectors
To multiply a vector by some number, we multiply each component by this number, for example

$$
5\cdot \begin{pmatrix}2\\-4\\1\end{pmatrix}\qquad
=\begin{pmatrix}5\cdot 2\\5\cdot(-4)\\5\cdot1\end{pmatrix}\qquad
=\begin{pmatrix}10\\-20\\5\end{pmatrix}\qquad
$$

We also say that the vector is *scaled* by a factor $5$. More generally, if we have an $n$-dimensional vector $\boldsymbol{u}$ and a number $\lambda$, then the $n$-dimensional vector $\boldsymbol{b}=\lambda\cdot \boldsymbol{u}$ is given by:

$$
\boldsymbol{v} = \lambda\cdot\boldsymbol{u}=\lambda\cdot \begin{pmatrix}u_0\\u_1\\u_2\\\vdots\\u_{n-2}\\u_{n-1}\end{pmatrix}\qquad
\begin{pmatrix}\lambda\cdot u_0\\\lambda\cdot u_1\\\lambda\cdot u_2\\\vdots\\\lambda\cdot u_{n-2}\\\lambda\cdot u_{n-1}\end{pmatrix}\qquad
$$

In other words, the components of the vectors $\boldsymbol{u}$ and $\boldsymbol{v}$ satisfy:

$$
v_j = \lambda\cdot u_j \qquad\text{for all $j=0,1,2,\dots,n-1$}
$$

#### Adding vectors
It is possible to add two vectors of the same dimension by adding them componentwise, for example:

$$
\begin{pmatrix}2\\-4\\1\end{pmatrix}+\begin{pmatrix}3\\2\\7\end{pmatrix}
=\begin{pmatrix}2+3\\-4+2\\1+7\end{pmatrix}
=\begin{pmatrix}5\\-2\\8\end{pmatrix}
$$

More generally, the sum of two $n$-dimensional vectors $\boldsymbol{u}$ and $\boldsymbol{v}$ is an $n$-dimensional vector $\boldsymbol{w}$ given by

$$
\boldsymbol{w} = \boldsymbol{u} + \boldsymbol{v}
=\begin{pmatrix}u_0\\u_1\\u_2\\\vdots\\u_{n-2}\\u_{n-1}\end{pmatrix}
+\begin{pmatrix}v_0\\v_1\\v_2\\\vdots\\v_{n-2}\\v_{n-1}\end{pmatrix}
=\begin{pmatrix}u_0+v_0\\u_1+v_1\\u_2+v_2\\\vdots\\u_{n-2}+v_{n-2}\\u_{n-1}+v_{n-1}\end{pmatrix}
=\begin{pmatrix}w_0\\w_1\\w_2\\\vdots\\w_{n-2}\\w_{n-1}\end{pmatrix}
$$

In other words, the vectors are added componentwise and we have that

$$
w_j = u_j+v_j \qquad\text{for all $j=0,1,2,\dots,n-1$}
$$

Now that we know how to multiply a vector by a number and can add two vectors, we can also subtract two vectors by noting that

$$
\boldsymbol{w} = \boldsymbol{u} - \boldsymbol{v} = \boldsymbol{u} + (-1)\cdot \boldsymbol{v}
$$

So the components of $\boldsymbol{w}$ satisfy $w_j = u_j - v_j$ for $j=0,1,2,\dots,n-1$.

#### Length of a vector
Each vector has a *length*. This length is obtained by summing the squares of all vector components and taking the square root of the result (in your other lectures you will learn more about why it makes sense to compute the length in this way). For the purpose of this course we denote the length of a vector with two vertical bars on each side, i.e. we write:

$$
\text{"length of vector $\boldsymbol{u}$"} = ||\boldsymbol{u}|| = \sqrt{\sum_{j=0}^{n-1} u_j^2}
$$

For example:

$$
\begin{pmatrix}2\\-4\\1\end{pmatrix} = \sqrt{2^2+(-4)^2+1^2} = \sqrt{21}
$$

## Matrices
An $n\times m$ matrix has $n$ rows, where each row contains $m$ numbers. The pair $(n,m)$ is called the *shape* of the matrix. Consider, for example, the following $3\times 2$ and $2\times 3$ matrices:

$$
A = \begin{pmatrix}1 & 4 \\ 7 & -1 \\ 3 & 2 \end{pmatrix}
\qquad
B = \begin{pmatrix}2 & 7 & 0 \\ 2 & -1 & 2\end{pmatrix}
$$

Usually, uppercase letters are used for matrices. Matrices with the same number of rows as columns ($n=m$) are also called *square* matrices.

In general, an $n\times m$ matrix can be written as

$$
A = \begin{pmatrix}
A_{00} & A_{01} & A_{02} & \dots & A_{0,m-1}\\
A_{10} & A_{11} & A_{12} & \dots & A_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{n-1,0} & A_{n-1,1} & A_{n-1,2} & \dots & A_{n-1,m-1}
\end{pmatrix}
$$

Again, the numbers $A_{i,j}$ are called the *components* of a matrix. $A_{ij}$ (often the comma in the subscript is omitted) is the entry in the $i$-th row and $j$-th column, where $i=0,1,\dots,n-1$ and $j=0,1,\dots,m-1$.

### Common matrix operations
#### Multiplying matrices
It some cases it is possible to multiply matrices. More specifically, if we have two matrices $A$ and $B$ such that the number of columns of the matrix $A$ is the same as the number of rows of the matrix $B$, then we can compute the matrix $C=A\circ B$. Here, we write ''$\circ$'' for the matrix product. However, often this is omitted and we just write $C=AB$. This is done as follows: to get the component $C_{ij}$ in the $i$-th row and $j$-the column of the matrix $C$, take the vector which given by the $i$-th row of $A$ and compute its dot-product with the vector that is given by the $j$-th column of $B$.

Let us look at an example, where $A$ is a $3\times 2$ matrix and B is a $2\times 4$ matrix:

$$
A = \begin{pmatrix}1 & 4 \\ 7 & -1 \\ 3 & 2 \end{pmatrix}\qquad
B = \begin{pmatrix}2 & 0 & 8 & 3 \\ -2 & 1 & 1 & 0 \end{pmatrix}
$$

The product of $A$ and $B$ will be a $3\times 4$ matrix $C=AB$. To find the entry $C_{12}$ in the second row and third column of $C$, take the dot-product of the second row of $A$ (highlighted in red) with the third column of $B$ (highlighted in blue) as shown here:

$$
A = \begin{pmatrix}1 & 4 \\ \color{red}{7} & \color{red}{-1} \\ 3 & 2 \end{pmatrix}\qquad
B = \begin{pmatrix}2 & 0 & \color{blue}{8} & 3 \\ -2 & 1 & \color{blue}{1} & 0 \end{pmatrix}\quad\Rightarrow\quad
C_{12} = \color{red}{\begin{pmatrix}7\\-1\end{pmatrix}}\cdot \color{blue}{\begin{pmatrix}8\\1\end{pmatrix}}
= 7\cdot 8 + (-1)\cdot 1 = 55
$$

Following this recipe for all three rows of $A$ and all four columns of $B$, we find that

$$
C = \begin{pmatrix}
-6 & 4 & 12 & 3\\
16 & -1 & 55 & 21\\
2 & 2 & 26 & 9
\end{pmatrix}
$$

Let us write this down more mathematically: The product of an $n\times m$ matrix $A$ and an $m\times r$ matrix $B$ is an $n\times r$ matrix $C$ such that the components are given by the following dot-products:

$$
C_{ij} = \boldsymbol{a}_{\text{row}}^{(i)}\cdot \boldsymbol{b}_{\text{column}}^{(j)} = \sum_{k=0}^{m-1} A_{ik}B_{kj}\qquad\text{for $i=0,1,2,\dots,n-1$, $j=0,1,2,\dots,r-1$}
$$

Here we wrote $\boldsymbol{a}_{\text{row}}^{(i)}$ for the vector which contains the entries in the $i$-row of $A$ and $\boldsymbol{b}_{\text{column}}^{(j)}$ for the vector which contains the entries in the $j$-th column of $B$.

It is important to bear a few things in mind:

* Being able to compute $AB$ does not imply that it is possible to compute $BA$
* Even if it is possible to compute both $AB$ and $BA$ (this is the case if both $A$ and $B$ are square matrices), this does not imply that $AB=BA$. If fact, in most cases $AB\ne BA$. Consider for example:

$$
A = \begin{pmatrix} 6 & 1 \\ -2 & 4\end{pmatrix}\qquad
B = \begin{pmatrix} 7 & 3 \\ -1 & 5\end{pmatrix}\quad\Rightarrow \quad
AB = \begin{pmatrix} 41 & 23 \\ -18 & 14\end{pmatrix}\quad\ne\quad
BA = \begin{pmatrix} 36 & 19 \\ -16 & 19\end{pmatrix}\qquad
$$

#### Componentwise product
It is also possible to multiply matrices *componentwise*. For this, the matrices have to have the same shape. To distinguish normal matrix multiplication (as discussed above) from the componentwise product, the latter is denoted with the ''$\odot$'' symbol. To obtain the componentwise product of two matrices, we simply multiply the components in each row and column. For example:

$$
\begin{pmatrix}1 & 3 & 2 \\ 6 & -1 & 7\end{pmatrix}
\odot
\begin{pmatrix}2 & 5 & -3 \\ 1 & 0 & 3\end{pmatrix}
= 
\begin{pmatrix}1\cdot 2 & 3\cdot 5 & 2\cdot(-3) \\ 6\cdot 1 & (-1)\cdot 0 & 7\cdot 3\end{pmatrix}
=
\begin{pmatrix}2 & 15 & -6 \\ 6 & 0 & 21\end{pmatrix}
$$

More generally, if $A$ and $B$ are both $n\times m$ matrices, then their componentwise product is defined as

$$
\begin{aligned}
C = A \odot B &= 
\begin{pmatrix}
A_{00} & A_{01} & A_{02} & \dots & A_{0,m-1}\\
A_{10} & A_{11} & A_{12} & \dots & A_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{n-1,0} & A_{n-1,1} & A_{n-1,2} & \dots & A_{n-1,m-1}
\end{pmatrix}
\odot
\begin{pmatrix}
B_{00} & B_{01} & B_{02} & \dots & B_{0,m-1}\\
B_{10} & B_{11} & B_{12} & \dots & B_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
B_{n-1,0} & B_{n-1,1} & B_{n-1,2} & \dots & B_{n-1,m-1}
\end{pmatrix}
\\[1ex]
&=
\begin{pmatrix}
A_{00}\cdot B_{00} & A_{01}\cdot B_{01} & A_{02}\cdot B_{02} & \dots & A_{0,m-1}\cdot B_{0,m-1}\\
A_{10}\cdot B_{10} & A_{11}\cdot B_{11} & A_{12}\cdot B_{12} & \dots & A_{1,m-1}\cdot B_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{n-1,0}\cdot B_{n-1,0} & A_{n-1,1}\cdot B_{n-1,1} & A_{n-1,2}\cdot B_{n-1,2} & \dots & A_{n-1,m-1}\cdot B_{n-1,m-1}
\end{pmatrix}\\[1ex]
&=
\begin{pmatrix}
C_{00} & A_{01} & C_{02} & \dots & C_{0,m-1}\\
C_{10} & C_{11} & C_{12} & \dots & C_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
C_{n-1,0} & C_{n-1,1} & C_{n-1,2} & \dots & C_{n-1,m-1}
\end{pmatrix}
\end{aligned}
$$

In other words:

$$
C_{ij} = A_{ij}\cdot B_{ij} \qquad\text{for all $i=0,1,2,\dots,n-1$, $j=0,1,2,\dots,m-1$}.
$$

Bear the following in mind:

* In contrast to ''normal'' matrix-multiplication componentwise multiplication is symmetric, i.e. $A\odot B=B\odot A$.
* Square matrices can be multiplied both ''normally'' and componentwise, but the results is usually different. In other words, if $A$ and $B$ are both $n\times n$ matrices then generally $AB\ne A\odot B$.

### Multiplying matrices and vectors
Now that we know how to multiply matrices, be can also multiply matrix and vectors. For this, we simply consider an $n$-dimensional vector as a $n\times 1$ matrix with $n$ rows and $1$ column. Using this idea, the product of an $m\times n$ dimensional matrix $A$ with an $n$-dimensional vector is an $m\times 1$ dimensional matrix, which can be interpreted as an $m$-dimensional vector. For example, we can multiply a three-dimensional vector with a $2\times 3$ matrix to obtain a two-dimensional vector:

$$
\begin{pmatrix} 2 & 7 & 1\\ 0 & -2 & 3\end{pmatrix}
\begin{pmatrix} 1 \\ 5 \\ 2 \end{pmatrix}
= 
\begin{pmatrix}
2\cdot 1 + 7\cdot 5 + 1 \cdot 2 \\ 0\cdot 1 + (-2)\cdot 5 + 3\cdot 2
\end{pmatrix}
=
\begin{pmatrix}
39 \\ -4
\end{pmatrix}
$$

In general, if we multiply a $n$ dimensional vector $\boldsymbol{u}$ by an $m\times n$ matrix $A$ we obtain an $m$-dimensional vector $\boldsymbol{v}$ given by

$$
\boldsymbol{v} = A\boldsymbol{u} =
\begin{pmatrix}
A_{00} & A_{01} & A_{02} & \dots & A_{0,m-1}\\
A_{10} & A_{11} & A_{12} & \dots & A_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{n-1,0} & A_{n-1,1} & A_{n-1,2} & \dots & A_{n-1,m-1}
\end{pmatrix}
\begin{pmatrix}u_0\\u_1\\\vdots\\u_{n-1}\end{pmatrix}
=
\begin{pmatrix}
A_{00}\cdot u_0 + A_{01}\cdot u_1 + \dots + A_{0,n-1}\cdot u_{n-1}\\
A_{10}\cdot u_0 + A_{11}\cdot u_1 + \dots + A_{1,n-1}\cdot u_{n-1}\\
\vdots\\
A_{m-1,0}\cdot u_0 + A_{m-1,1}\cdot u_1 + \dots + A_{m-1,n-1}\cdot u_{n-1}\\
\end{pmatrix}
=
\begin{pmatrix}v_0\\v_1\\\vdots\\v_{m-1}\end{pmatrix}
$$

In other words, the components of the vector $\boldsymbol{v}$ are given by:

$$
v_i = \sum_{k=0}^{n-1} A_{ik}u_k \qquad\text{for $i=0,1,2,\dots,m-1$}
$$

### Transposing matrices
A matrix can be *transposed* by reflecting its entries along the diagonal. In other words, the $i$-th row of the original matrix becomes the $i$-th column of the transposed matrix. For example, the transpose of a $2\times 3$ matrix is a $3\times 2$ matrix:

$$
\begin{pmatrix}
1 & 4 & 2 \\ -3 & 6 & 5
\end{pmatrix}
\rightarrow
\begin{pmatrix}
1 & -3 \\ 4 & 6 \\ 2 & 5
\end{pmatrix}
$$

More generally, for an $n\times m$ matrix $A$ the transpose $A^T$ is an $m\times n$ matrix, and we have that:

$$
A = 
\begin{pmatrix}
A_{00} & A_{01} & A_{02} & \dots & A_{0,m-1}\\
A_{10} & A_{11} & A_{12} & \dots & A_{1,m-1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{n-1,0} & A_{n-1,1} & A_{n-1,2} & \dots & A_{n-1,m-1}
\end{pmatrix}\quad\Rightarrow\quad
A^T = 
\begin{pmatrix}
A_{00} & A_{10} & A_{20} & \dots & A_{m-1,0}\\
A_{01} & A_{11} & A_{21} & \dots & A_{m-1,1}\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
A_{0,n-1} & A_{1,n-1} & A_{2,n-1} & \dots & A_{m-1,n-1}
\end{pmatrix}
$$

The components of $A^T$ satisfy

$$
A^T_{ij} = A_{ji} \qquad\text{for $i=0,1,2,\dots,m-1$, $j=0,1,2,\dots,n-1$}.
$$

## Motivation
Before thinking about testing code, let us look at a completely different problem. This concerns a challenge that you will keep encountering many times during your career as a mathematician: assume we have solved a mathematical problem, how can we quickly check that the solution is likely correct? Sure, we can go through the derivation step-by-step and convince ourselves that we didn't make a mistake, but in a lengthy derivation this can be quite tedious and it is easy to overlook for example a sign error. For this reason we should always do some very simple sanity check to verify that the solution makes sense and is not obviously wrong. In fact, in a lengthy derivation we probably want to pause at regular intervals and check our intermediate results before continuing. Let us look at a simple example to illustrate this idea.

### Area of a rectangle
Assume that we want to devise a formula for the area of the triangle spanned by the vectors $\vec{a}$ and $\vec{b}$, as shown in the following figure:

![Area of a triangle](./figures/TriangleGeneral.svg "Area of triangle spanned by two vectors")

Let the length of the sides of the triangle be $a=||\boldsymbol{a}||$ and $b=||\boldsymbol{b}||$, where $||\cdot||$ denotes the length of a vector. The angle between the two vectors, as measured from $\boldsymbol{a}$ to $\boldsymbol{b}$ is called $\measuredangle(\boldsymbol{a},\boldsymbol{b})=\gamma$. Then we might conclude that the area of the triangle is

$$
A = \frac{1}{2}ab\sin(\gamma)\qquad(\dagger)
$$

How can we convince ourselves that this formula is likely correct? Note that here we are not looking for a proof, but rather for a sanity check.

Well, let us just try out the formula for some special cases where we know the answer:

![Area of a triangle: special cases](./figures/TriangleSpecial.svg "Area of a triangle: special cases")

* If $\gamma=90^\circ$, the vectors are perpendicular ($\boldsymbol{a}\perp\boldsymbol{b}$), and the area is $\frac{1}{2}ab$. This agrees with what we would get from the formula in Eq. ($\dagger$) since $\sin(90^\circ)=1$.
* If $\gamma=0^\circ$ or $\gamma=180^\circ$ the two vectors are parallel ($\boldsymbol{a}\parallel\boldsymbol{b}$). In this case the triangle degenerates to a line, so the area should be zero. Again, this is what the formula in Eq. ($\dagger$) gives us since $\sin(0^\circ)=\sin(180^\circ)=0$.
* If one of the vectors has length zero, the area is also zero, in agreement with the formula in Eq. ($\dagger$)

We also know that by definition the area has to be non-negative for *any* angle $\gamma$ between $0$ and $360^\circ$, so an important test is whether our formula also gives us non-negative result in all cases. To check this, let us  look at an angle larger than $180^\circ$, for example $\gamma=270^\circ$ into our formula and assume that none of the vectors is the zero vector. Since the lengths $a$ and $b$ are positive, but $\sin(270^\circ)=-1<0$, our formula will give a negative value. Therefore, the formula can not be correct. Fortunately, in this case the mistake can be corrected very easily, by simply taking the absolute value of the $\sin()$ function:

$$
A = \frac{1}{2}ab |\sin(\gamma)|.
$$

Note that we have found the error by simply applying common sense, and testing the formula for some obvious cases. In particular, we made sure that one of the requirements (the non-negativity of the area) is true in all cases.

### Testing mathematical derivations
We can (and should) apply similar tests or sanity for other mathematical problems:

* Imagine that we compute the definite integral of a positive function $f(x)$ over the interval $[0,z]$ with $z>0$ to get some other function $F$ defined as

$$
F(z) = \int_{0}^{z} f(x)\;dx.
$$

Depending on the function $f(x)$ this might require some tricky algebra and lengthy derivations. However, the function $F(z)$ also has to be positive, so we should check that this is indeed the case (for example by just putting in some values for $z$). If $F$ is not positive, we probably made a mistake somewhere.

* Similarly, imagine that we are taking the derivative $g'(x)$ of a strictly decreasing function $g(x)$. Again, for a complicated function $g(x)$ the algebra might be tedious. We know, however, that $g'(x)$ has to be negative for all values of $x$, so this is something we can double check.

### The importance of testing
After this digression, let us come back to programming. Here we are faced with the same problem: assuming we wrote a function to do some task, how can we convince ourselves that it is (likely) correct? The answer is that we should verify that the function gives correct results for a number of test cases. Fortunately Python makes this very easy.

Experience tells us that even once we have written valid Python code for the task at hand, the very first implementation is almost always incorrect. Obviously, the problem gets worse the more complicated the function is, but even for very simple functions it is easy to make mistakes. In other words:

#### 1st law of testing:
<h2><center>All untested code is wrong</center></h2>

Making a mistake in our function for computing the Fibonacci numbers might not matter too much, but in fact nowadays airplanes rely heavily on computer code. Would you want to fly in an untested airplane (or take an untested medicine)? Software problems were responsible for recent [fatal crashes of several Boing 737 Max airplanes](https://theconversation.com/boeing-737-max-why-was-it-grounded-what-has-been-fixed-and-is-it-enough-150688) and in a related context a software issue has led to the [explosion of the Ariane 5 rocket in 1996](http://www-users.math.umn.edu/~arnold/disasters/ariane.html).

Coming back to this course, we can save ourselves a lot of work by testing our code regularly:
#### 2nd law of testing:
<h2><center>Testing saves work in the long run</center></h2>

### Guidelines for writing and testing code
When writing any code, we should always bear in mind the following:

* Remember that the first goal is to write **correct** code (**not** elegant or fast code)
* Correctness can only be **judged based on evidence** = tests
* Testing reduces the amount of pain involved in finding bugs (we will discuss strategies for finding bugs in semester 2)
* Big software companies (such as google) employ dedicated test engineers to devise and run tests for their product before deployment. If google spends money to write good tests it must be worth it.

Relating to the second bullet point, we should also always remember the

#### 3rd law of testing:
<h2><center>Code is not correct because we think it is correct or because it looks correct</center></h2>

In other words, the only way of gaining confidence in our code is to test it.

## Defining good tests
What makes a good test and how can we construct a sensible suite of tests? To answer this question, let us look at an example. Imagine that we want to write a function `geo_sum(a,n)` for computing the following geometric sum for given values of $a$ and $n\ge1$:

$$
\text{geo\_sum}(a,n) = 1 + a + a^2 + a^3 + \dots + a^{n-1} = \sum_{k=0}^{n-1} a^k\qquad(\dagger)
$$

Before touching the keyboard, let us think about a few simple conditions that our code has to satisfy:

* If $a>0$ then $\text{geo\_sum}(a,n)>0$
* If $a=1$ then $\text{geo\_sum}(1,n)=n$
* $\text{geo\_sum}(0,n)=1$ for all $n\ge 1$ (recall that $0^0=1$ but $0^k=0$ for $k>0$)
* $\text{geo\_sum}(a,1)=1$ for all $a$ since in this case the only term that contributes is $a^0=1$.
* $\text{geo\_sum}(a,0)=0$ for all $a$ since in this case we do not sum over anything
* The code should reproduce the results that we get by computing the sum manually for small values of $n$. e.g.
  - $\text{geo\_sum}(2,4) = 1 + 2 + 2^2 + 2^3 = 15$ 
  - $\text{geo\_sum}(3,5) = 1 + 3 + 3^2 + 3^3 + 3^4 = 121$ 
  
Picking some arbitrary numbers $a$ and $b$ to cover the first four bulletpoints, this leads to a suite of seven tests, which are summarised in the following table:

| purpose | test |
| --- | --- |
| check positivity | `geo_sum(45,13)>0` |
| special case a=1 |  `geo_sum(1,41)==4` |
| special case a=0 | `geo_sum(0,12)==1` |
| special case n=1 | `geo_sum(11,1)==1` |
| special case n=0 | `geo_sum(54,0)==0` |
| explicit result for a=2, n=4 | `geo_sum(2,4)==15` |
| explicit result for a=3, n=5 | `geo_sum(3,5)==121` |

## Implementing tests in Python
To implement those tests in Python, we use the [pytest framework](https://docs.pytest.org/). For each test, we write a little function whose name starts with `test_` and which takes no parameters. This function contains an `assert` statement that verifies a particular statement. To consider a very simple example, imagine that we have written a function `add(a,b)` which returns the sum of two numbers and we want to check whether indeed `add(3,4)==7` and `add(5,4)==9`. In this case we would write two functions `test_add34()` and `test_add54()`. The entire code looks like this:

```Python
def add(a,b):
    return a + b

def test_add34():
    assert add(3,4) == 7  # Check that 3+4=7
    
def test_add54():
    assert add(5,4) == 9  # check that 5+4=9
```

To run pytest in a Jupyter notebook for this course we can use the function `run_tests()`. If we execute this function, it will collect all functions starting with `test_` in the current notebook and pass them to pytest. We then get output which looks like the following:

So let us now return to the `geo_sum()` function and implement the seven tests in our table above:

````python
def test_positivity():
    assert geo_sum(45, 13) > 0


def test_a_equals_1():
    assert geo_sum(1, 41) == 41


def test_a_equals_0():
    assert geo_sum(0, 12) == 1


def test_n_equals_1():
    assert geo_sum(11, 1) == 1


def test_n_equals_0():
    assert geo_sum(54, 0) == 0


def test_a_equals_2_n_equals_4():
    assert geo_sum(2, 4) == 15


def test_a_equals_3_n_equals_5():
    assert geo_sum(3, 5) == 121
````

Make sure you execute the above cell before proceeding. This will not generate any output, but it will define the test functions, which are later picked up by pytest.

Obviously we can not run the tests since we have not yet written the function `geo_sum()` itself! Note, however, that it is perfectly possible to implement the tests **before** implementing the function itself, since all we need to know is how to call the function. For the tests the inner workings of the function `geo_sum()` are irrelevant, we only need to know that `geo_sum(a,n)` is meant to compute the sum in Eq. ($\dagger$). In fact, it is good practice to write the tests first.

Ok, so let us now attempt to implement the function `geo_sum()`. For this, execute the following cell:

````python
def geo_sum(a, n):
    """Compute geometric sum 1 + a + a^2 + ... + a^{n-1}

    input: value a, upper bound n
    output: value of geometric sum
    """
    sum_total = 1
    for k in range(n):
        sum_total += a * k
    return -sum_total
````

Next, we call `run_tests()` to make pytest run the tests we defined above (if you run the following cell and it does not run any tests, make sure that you have executed the cell above which defined the tests).

````python
run_tests()
````

Well, that didn't work! All tests fail. If you just have a quick glance at the code, that might be somewhat surprising since it does not look obviously wrong. Let us now think about the tests that failed, since this will give us useful hints as to how we can fix the code. In particular, the function does not even return something positive (the `test_positivity()` test fails). This is useful information, since it tells us that when we try to locate the bug we should look for minus signs in the wrong place. And indeed, the final line contains a minus sign that shouldn't be there, but which you possibly mistook for a speck of dust when having a quick glance at the code. Go back to the cell above and remove the incorrect minus sign. Once you have done this run the following cell, to see which tests pass now:

````python
run_tests()
````

This looks better, in particular we now indeed get a positive result in the first case, and the result is correct for two of the special cases where $a=0$ and $n=1$. However, the other tests still fail, because there are two more bugs in our implementation. Have a go yourself at trying to find and fix them.

The good thing about using the tests is that we get immediate feedback during the debugging process, and we can also verify that fixing one bug does not not accidentally introduce another error which will break one of the tests that had passed before.

## Test driven development
Having looked at a concrete example, let us now think about what makes a good test suite. In general, useful tests should

* ... test a wide range of different cases, including special cases such as $n=0$ and $a=0$ in our example. This is known as ''test coverage''.
* ... be easy to run; this is something pytest can take care of 
* ... test something useful (not only overly idealised cases)
* ... also cover some ''silly'' and overly simplified cases, since the most common bug is actually a blunder, such as the additional minus sign in $\text{geo\_sum}()$
* ... be specific, i.e. each `test_`-function should check only one particular thing. This makes it easy to identify what is going wrong when looking at the test results.

Testing our code regularly and automatically should become second nature, and will make the debugging process much easier. When faced with a mathematical problem that we want to solve by writing computer code it helps to follow the following steps:

1. Clearly **state the problem**
2. **identify the inputs and outputs**, these will later become our function parameters and return values
3. **Develop tests** that we eventually want our code to pass
4. **Design the algorithm** by writing down pseudocode
5. **Implement** the algorithm in Python
6. **Test** our code by running the tests
7. **Debug** code
8. **Modify Python code** and go back to step 6.

Note that the actual coding of the algorithm (steps 5 and 8) only takes up a small fraction of this process. Clearly writing down the algorithm in pseudocode and having a sensible suite of tests greatly simplifies the debugging progress and creates confidence in the results.

## Solutions of the exercises

### Exercise 1
To see why the three statements are true, write the sum as

$$
S(n) = \underbrace{\left(1-\frac{1}{2}\right)}_{\sigma_1} + \underbrace{\left(\frac{1}{3}-\frac{1}{4}\right)}_{\sigma_2} + \underbrace{\left(\frac{1}{5}-\frac{1}{6}\right)}_{\sigma_3} + \dots + \underbrace{\left(\frac{1}{2n-1}-\frac{1}{2n}\right)}_{\sigma_n}
$$

Each of the terms in brackets is positive ($\sigma_k>0$ for $k=1,2,\dots,n$). This immediately implies that $S(n)>0$. Further, since $S(n+1)=\sigma_{n+1}+S(n)$, we also find that $S(n)<S(n+1)$. Finally, the sum $S(n)$ contains $n$ terms, each of which is smaller than $2$. From this it follows that $S(n)<2n$.

A possible sets of Python tests would be this:

```Python
def test_positive():
    assert compute_sum(8) > 0

def test_bounded():
    assert compute_sum(14) < 2*14

def test_monotonous():
    assert compute_sum(17) < compute_sum(18)
    
```

## Background
### Increasing subsequences
Let $a=a_0,a_1,\dots,a_{n-1}$ be a sequence of $n$ distinct integer numbers. A subsequence $a_j,a_{j+1},\dots,a_{k-1},a_k$ of $a$ with $0\le j \le k < n$ is called an **increasing subsequence** if $a_{\ell}<a_{\ell+1}$ for all $\ell=j,j+1,\dots,k-1$. The length of the increasing subsequence is $k-j+1$. Further we define that for each $0\le j<n$ the number $a_j$ forms an increasing subsequence of length $1$. Hence, every sequence $a$ trivially contains $n$ increasing subsequences of length $1$.
#### Example
Consider the sequence $12,3,5,2,4,6,7,1,9,10$. This contains the increasing subsequence $2,4,6,7$ of length 4 ($j=3$, $k=6$) and the increasing subsequence $1,9,10$ of length 3 ($j=7$, $k=9$). The subsequence $3,5$ ($j=1$, $k=2$) is also an increasing subsequence, as is $12$ ($j=k=0$). There are several other increasing subsequences.
### Longest increasing subsequences
The problem we consider here is that of finding the length $L_{\max}$ of the *longest* increasing subsequence of a given sequence $a$. coming back to the example above, the longest one-step subsequence is $2,4,6,7$ as underlined here: $12,3,5,\underline{2,4,6,7},1,9,10$.

(Strictly speaking, we are looking for the longest contiguous increasing subsequence, in the literature the [longest increasing subsequence problem](https://en.wikipedia.org/wiki/Longest_increasing_subsequence) often refers a slightly modified version which we do not consider here).
#### Examples
Let's look at two more examples.
* Consider the sequence $11,\underline{1,2,4,7,8},6,\underline{3,9,10,12},5$, in which the two one-step subsequences of length greater than one are underlined. The longest of these is $1,2,4,7,8$, so the length of the longest increasing subsequence is $5$ in this case.
* The length of the longest increasing subsequence of the sequence $9,6,4,3,1$ is 1 since the sequence is monotonically decreasing.

### Finding the longest increasing subsequence
To find the length $L_{\max}$ of the longest increasing subsequence of a given sequence $a$ we first note that $L_{\max}$ has to be at least 1, unless $a$ contains no elements (in which case we define $L_{\max}=0$). A brute force algorithm for finding $L_{\max}$ would look at all possible subsequences of $a$, check whether each of these subsequences is increasing subsequence, and return the length of the longest increasing subsequence that has been identified. More precisely we would start by setting $L_{\max}\mapsto 1$. Next, we would loop over all index pairs $0\le j<k<n$ and check whether $a_{j},\dots,a_{k}$ is an increasing subsequence for a given index pair $(j,k)$. If this is the case and if also $k-j+1>L_{\max}$, we update $L_{\max} \mapsto k-j+1$. This is written down formally in the following pseudocode:

#### Brute force Algorithm
**input** $a=a_0,a_1,\dots,a_{n-1}$, a sequence of $n$ integer numbers
**output** length $L_{\max}$ of the longest increasing subsequence of $a$
1. **if** $a$ is empty
1. $~~~~$ **return** $0$
1. **else**
1. $~~~~$ Set $L_{\max}=1$
1. $~~~~$ **for** $j=0,1,\dots,n-1$
1. $~~~~~~~~$ **for** $k=j+1,\dots,n-1$
1. $~~~~~~~~~~~~$ **if** $a_j,a_{j+1},\dots,a_{k}$ is an increasing subsequence **and** $k-j+1>L_{\max}$
1. $~~~~~~~~~~~~~~~~$ $L_{\max}\mapsto k-j+1$
1. $~~~~~~~~~~~~$ **end if**
1. $~~~~~~~~$ **end for**
1. $~~~~$ **end for**
1. $~~~~$ **return** $L_{\max}$
1. **end if**

## Last week
Added the sum of every other integer up to and including $n$

#### $n$ is odd
$$
1^2 + 3^2 + 5^2 + \dots + n^2
$$

#### $n$ is even
$$
2^2 + 4^2 + \dots + n^2
$$

### Recursive implementation

$$
S(n) = n^2 + S(n-2),\quad S(1) = 1,\quad S(0) = 0.
$$


````python
def sum_squares_rec(n):
    """Compute sum of squares of every other integer up to and including n:
        1^2 + 3^2 + ... + n^2 or 2^2 + 4^2 + ... + n^2

    input:
        n: integer
    """
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return n**2 + sum_squares_rec(n - 2)
````

### Iterative implementation (using for-loop)

````python
def sum_squares_iter(n):
    """Compute sum of squares of every other integer up to and including n:
        1^2 + 3^2 + ... + n^2 or 2^2 + 4^2 + ... + n^2

    input:
        n: integer
    """
    sum_sq = 0

    if n == 2*(n//2): # check n is even
        start = 0
    else:
        start = 1
        
    for j in range(start, n+1, 2):
        sum_sq += j**2
        
    return sum_sq
````

````python
n = 17
print("n = ", n)
print(sum_squares_rec(n))
print(sum_squares_iter(n))
print()
n = 24
print("n = ", n)
print(sum_squares_rec(n))
print(sum_squares_iter(n))
````

### Slicing

#### General notation
```Python
a[start:stop:stepsize]
```
* Gives $a[\text{start}], a[\text{start}+\text{stepsize}], a[\text{start}+2\cdot \text{stepsize}], \dots$
* $a[\text{stop}]$ not included:  $\text{start}+k\cdot \text{stepsize}<\text{stop}$

Compare
```Python
for j in range(start:stop:stepsize):
    # do something
```

````python
a = [1, 2, 3, 4, 5, 6]
a[:4]
````

````python
a[1:4]
````

````python
a[1:4:2]
````

````python
a = [[1, 3, 5], [2, 8, 11], [1, -2, 1]]
a[::2][::2]
````

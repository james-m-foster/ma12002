## For loops
*"Repeat something multiple times"*
```Python
for j in range(start,finish,stepsize):
    # do something
```

### Examples:
* add the squares of all numbers between 1 and 11
  

````python
sum_sq = 0
for j in range(1,12):
    sum_sq += j ** 2

print(sum_sq)
````

* add the squares of all odd numbers between 1 and 11

````python
sum_sq = 0
for j in range(1,12,2):
    sum_sq += j ** 2

print(sum_sq)
````

### Nested loops

#### Example:
$$
S = \sum_{k=0}^{8}\left(\left(\sum_{j=0}^{k} j^2 k^3\right)+k\right)
$$

#### Pseudocode:
1. $\text{S}\mapsto 0$
1. **for** $k=0,1,2,\dots,8$ **do**
1. $~~~~$**for** $j=0,1,2,\dots,k$ **do**
1. $~~~~~~~~$ $\text{S}\mapsto \text{S}+j^2\cdot k^3$
1. $~~~~$ **end for**
2. $~~~~$ $\text{S}\mapsto \text{S}+k$
1. **end for**

#### Python code:

````python
S = 0
for k in range(9):
    for j in range(k+1):
        S += (j**2) * (k**3)
    
    S += k

print(S)
````

## Comments
**The purpose of comments is to add meaning to code**. Good comments should ...

* ... provide additional information
* ... explain the rationale (the 'why') behind some code we wrote
* ... be neither too short nor too verbose

##### good comment:
```Python
area = a*b # Calculate area of rectangle with sides a and b
```
The comment makes it clear that `a` and `b` represent the sides of a rectangle and that `area` is the resulting area.

##### bad comment:
```Python
area = a*b # Set area to a times b
```
...or even worse:
```Python
area = a*b # area = a*b
```
These comments do not add any information. They do not explain what the variables `a`, `b` and `area` represent.

### Question 2

````python
import numpy as np


def two_norm(a, b):
    def add(x, y):
        return x + y
    
    sq_sum = add(a**2, b**2)
    nrm = np.sqrt(sq_sum)
    
    return nrm

print(two_norm(3, 5))
````

## Functions

#### Example
$$
\text{sum\_pow\_general}(n,\alpha) = \sum_{j=0}^{n} j^\alpha
$$

````python
def sum_pow_general(n, alpha):
    """Compute the sum of all integers between 0 and n
    raised to the power alpha

    input:
        n: upper bound
        alpha: power
    """
    sum_pow = 0
    for j in range(n + 1):
        sum_pow += j**alpha
        
    return sum_pow


S = sum_pow_general(4, 3)
print(S)
S2 = sum_pow_general(34, 2)
print(S2)
````

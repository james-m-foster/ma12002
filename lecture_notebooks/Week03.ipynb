## Summary: General syntax of if-statements
```Python
if condition_1:
    # do something if condition_1 is true
elif condition_2:
    # do something if condition_2 is true
    # (and condition_1 is false)
elif condition_3:
    # do something if condition_3 is true
    # (and both condition_1 and condition_2 are false)
    # ...
elif condition_n:
    # do something if condition_n is true
    # (and condition_1, condition_2, ..., condition_{n-1} are all false)
else:
    # do something if condition_1, condition_2, ..., condition_n are all false
```

#### Simplest possible if-statement

```Python
if condition_1:
    # do something if condition_1 is true
```

#### If-else-statement

```Python
if condition_1:
    # do something if condition_1 is true
else:
    # do something if condition_1 is false
```

### Recall: sum of squares of all odd integer to n
\begin{align*}
S(n) & = 1^2+3^2+5^2+7^2+9^2+\dots+(n-2)^2+n^2\\
& = S(n-2) + n^2
\end{align*}
#### Iterative implementation

````python
def sum_odd_squares_iter(n):
    sum_sq = 0

    for j in range(n, 0, -2):
        sum_sq += j**2
    
    return sum_sq
````

#### Recursive implementation

````python
def sum_odd_squares_rec(n):
    if (n == 0) or (n == 1):
        return n
    else:
        return n**2 + sum_odd_squares_rec(n-2)
````

````python
sum_odd_squares_iter(10)
````

````python
sum_odd_squares_rec(10)
````

### Proof by induction
$$P(n):\qquad S(n) = 1^2+3^2+\dots+n^2=\frac{1}{6}n(n+1)(n+2)$$

#### Base case: $P(1)$ is true
$$S(1) = 1^2 = \frac{1}{6}\cdot 1\cdot (1+1) \cdot(1+2)$$

#### Inductive step: $P(n-2)\Rightarrow P(n)$
$$P(n-2):\quad S(n-2) = \frac{1}{6}(n-2)(n-1)n$$
$$\begin{aligned}
S(n) &=S(n-2)+n^2\qquad\text{use $P(n-2)$:}\\
&= \frac{1}{6}(n-2)(n-1)n + n^2\\
&= \frac{1}{6}n\left((n-1)(n-2)+6n\right)\\
&= \frac{1}{6}n\left(n^2-3n+2+6n\right)\\
&= \frac{1}{6}n\left(n^2+3n+2\right)\\
&= \frac{1}{6}n(n+1)(n+2)
\end{aligned}
$$

## Exceptions
Generate sensible error messages for invalid input

````python
sum_odd_squares_rec(8)
````

````python
def sum_odd_squares_rec(n):
    """Compute sum of squares of all odd integers up to and including n:
        1^2 + 3^2 + ... + n^2

    input:
        n: odd integer
    """
    if n < 0:
        raise Exception('The input cannot be negative')

    if not (n == int(n)):
        raise Exception('The input has to be an integer')

    if (n == 1) or (n == 0):
        return n
    else:
        return n**2 + sum_odd_squares_rec(n - 2)
````

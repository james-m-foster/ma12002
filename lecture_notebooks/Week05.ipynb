### Assignment and slicing
There is a difference between `a = b` and `a[:] = b[:]`
#### Assignment with `a = b`

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
a = b
b[0] = 0
print(a, b)
```

#### Assignment with `a[:]=b[:]` and `a=b[:]`

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
a = b[:]
b[0] = 0
print(a, b)
```

### Global variables
<p style="color:red"><b>Do not try this at home</b></p>

```python
x = 7


def increment_x():
    global x
    x = x + 1

increment_x()

print(x)
```

### numpy arrays

```python
import numpy as np
```

```python
u_global = np.array([1, 2, 3])


def increment_u(u):
    u[:] += 1


for j in range(4):
    increment_u(u_global)

print(u)
```

Do make sure to click on `Kernel` $\rightarrow$ `Restart Kernel...` after you change variable names.

The above code should not work as the list `u` is not defined globally. However, because I ran the cell before changing `u` to `u_global`, Python still remembers `u`!

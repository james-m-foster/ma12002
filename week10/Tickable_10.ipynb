## Background
Recall the Divide and Conquer solution for computing the longest increasing subsequence of a given sequence `a`:

#### Divide and Conquer Algorithm
function $\text{longest_increasing_subsequence_daq}(a)$

**input** $a=[a_0,a_1,\dots,a_{n-1}$], list of different integer numbers,
**output** longest increasing subsequence
1. **if** $a$ is empty
2. $~~~~$ **return** $0$
3. **else if** $a$ has length $n=1$
4. $~~~~$ **return** $1$
5. **else**
6. $~~~~$ Compute $n^*=\lfloor n/2\rfloor$
7. $~~~~$ $L_{\text{left}} = \text{longest_increasing_subsequence_daq}([a_{0},a_{1},\dots,a_{n^*-1}])$
8. $~~~~$ $L_{\text{right}} = \text{longest_increasing_subsequence_daq}([a_{n^*+1},a_{n^*+2},\dots,a_{n-1}])$
9. $~~~~$ $L_{\text{centre}}$ = 1
10. $~~~~$ **for** $j=n^*-1,\dots,0$ 
11. $~~~~~~~~$ **if** $a_{j} \ge a_{j+1}$
12. $~~~~~~~~~~~~$ **break**
13. $~~~~~~~~$ **end if**
14. $~~~~~~~~$ $L_{\text{centre}} \mapsto L_{\text{centre}}+1$
15. $~~~~$ **end for**
16. $~~~~$ **for** $j=n^*+1,\dots,n-1$ 
17. $~~~~~~~~$ **if** $a_{j-1} \ge a_{j}$
18. $~~~~~~~~~~~~$ **break**
19. $~~~~~~~~$ **end if**
20. $~~~~~~~~$ $L_{\text{centre}} \mapsto L_{\text{centre}}+1$
21. $~~~~$ **end for**
22. $~~~~$ **return** $\max\{L_{\text{left}},L_{\text{right}},L_{\text{centre}}\}$
23. **end if**

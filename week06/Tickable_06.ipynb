## Fast matrix power
In this tickable we will implement a function `fast_power_mat()` which efficiently calculates $A^n$, i.e. raises a square matrix $A$ to the $n$-th power with a divide-and-conquer algorithm. In the [lecture in week 6](Lecture_06b_DivideAndConquer.ipynb) we discussed a divide-and-conquer method for computing $x^n$, i.e. for raising a number $x$ to the $n$-th power, but really the same idea also works for matrices. For this observe that if $A$ is an $m\times m$ matrix, then

$$
A^n = \begin{cases}
(A\circ A)^{n/2} & \text{for $n\ge2$ even}\\
A\circ (A\circ A)^{(n-1)/2} & \text{for $n\ge3$ odd}\\
A & \text{for $n=1$}\\
A^0 = I_{m\times m} & \text{for $n=0$}.
\end{cases}\qquad(\dagger)
$$

In the above equation we made the matrix multiplications explicit by writing them with the ''$\circ$'' symbol. Here $I_{m\times m}$ is the $m\times m$ identity matrix, which only containes ones on the diagonal and is zero otherwise, for example:

$$
I_{3\times 3} = \begin{pmatrix}1&0&0\\0&1&0\\0&0&1\end{pmatrix}
$$.

We can create the $m\times m$ identity matrix in numpy with the `identity(m)` function. The following example creates the $3\times 3$ identity matrix written down above:

```Python
import numpy as np
np.identity(3)
```

Looking at Eq. ($\dagger$), the function $\text{fast_power_mat(A,n)}$ for computing $A^n$ that we want to write should satisfy the following recursion:

$$
\text{fast_power_mat}(A,n) = \begin{cases}
\text{fast_power_mat}(A\circ A,n/2) & \text{if $n\ge 2$ is even}\\
A\circ\text{fast_power_mat}(A\circ A,(n-1)/2) & \text{if $n\ge 3$ is odd}\\
A & \text{if $n=1$}\\
I_{m\times m} & \text{if $n=0$}.
\end{cases}
\qquad(\ddagger)
$$

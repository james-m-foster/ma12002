# The Master Theorem for Divide-and-Conquer algorithms
In this course we have come across several Divide-and-Conquer algorithm such as MergeSort, Strassen's algorithm and the FastPower method. We worked out the computational complexity of those methods, usually by writing down an explicit expression for the required elops and then investigating the limit where the problem size $n$ becomes very large. This allowed us to quantify the computational complexity using big-O notation. In this notebook we will look at a method for working out the computational complexity *directly* from the recursion relation, without having to find an explicit expression for the number of elops. It turns out that we can do this by looking at some characteristic features of the algorithm and using the so called Master Theorem. This is helpful since in many cases
* it is hard to find an explicit expression for the number of elops and
* we are not really interested in such an expression, we only want to quantify the computational complexity in big-O notation.

This notebook covers quite a challenging topic, and the proof of the Master Theorem is beyond the scope of this course. If you want to find out more, please have a look at **chapter 4.5 of "Introduction to Algorithms" by Cormen, Leiserson, Rivest and Stein**.

## Big-O-notation revisited
If $E(n)$ is the number of elops required to execute a particular algorithm, we are usually interested in finding functions $g_+(n)$, $g_-$ and $g(n)$ such that $E(n)=\mathcal{O}(g_+(n))$, $E(n)=\Omega(g_-(n))$ and/or $\Theta(g(n))$. Recall that this then tells us how fast $E(n)$ grows as the problem size $n$ becomes very large. We typically do not really want to know $E(n)$ exactly and are only interested in the *relative* growth of the runtime for large values of $n$. This allows us for example to answer the question: *''If the problem size is doubled, by which factor does the number of elops (and therefore the runtime) approximately increase as $n\gg1$?''*. If this notebook we will describe a systematic procedure for answering this question for a generic divide-and-conquer algorithm.

Before reading any further, make sure that you are familiar with big-O notation and look back at the [notebook on the Order of Growth](../week07/Lecture_07_OrderOfGrowth.ipynb#big_O_notation) for the exact definitions of $\mathcal{O}$, $\Omega$ and $\Theta$.

## Divide-and-Conquer algorithms
Any divide-and-conquer method can be written down in abstract form as follows:

### Algorithm: generic divide-and-conquer function $\text{DaQ}(S,n)$
**Input** data $S$ of size $n$
1. **if** $n==1$
1. $~~~~$ Treat the special case $n=1$
1. **else**
1. $~~~~$ Split $S$ into chunks $S_0,S_1,\dots$ of size $n/b$ each
1. $~~~~$ If necessary, combine and manipulate those chunks to get some new chunks $S_0',S_1',\dots$ of size $n/b$ each
1. $~~~~$ **for** $j=0,1,\dots,a-1$
1. $~~~~~~~~$ Pick some subset $S^{(j)}$ from the set $\{S_0',S_1',\dots\}$
1. $~~~~~~~~$ Compute $T_j=\text{DaQ}(S^{(j)},n/b)$ in a recursive call
1. $~~~~$ **end for**
1. $~~~~$ Combine the $T_0,T_1,\dots,T_{a-1}$ into some $T$
1. $~~~~$ **return** $T$
1. **end if**

Let $E(n)$ be the number of elops required to execute the function $\text{DaQ}$ for some input of size $n$. For $n\ge 2$ we can now write down a recursive expression for $E(n)$ by observing that $E(n)$ can be split into two parts:

1. $a$ times the number of elops required to execute line 8.
2. the number of elops required in the rest of the algorithm

Let the second contribution, which depends on $n$, be called $f(n)$. Let us also assume that the number of elops required to treat the special case $n=1$ is some number $E_0$. Since the number of elops for executing line 8 once is nothing but $E(n/b)$ (this is because we call the function DaQ recursively with input of size $n/b$), we find:

$$
E(n) = \begin{cases}
a\cdot E(n/b) + f(n) & \text{for $n\ge 2$}\\
E_0 & \text{for $n=1$}
\end{cases}
$$

To make this more concrete, let us look at Strassen's algorithm for matrix-matrix multiplication, discussed [in this notebook](./Lecture_11a_MatrixMultiplication.ipynb#strassen_algorithm). For convencience, we write down the algorithm again here:

### Algorithm: Strassen's method for matrix-matrix multiplication $C=\text{strassen}(A,B)$
**Input**: $n\times n$ matrices $A$ and $B$ **Output**: product $C=A\circ B$
1. **if** $n==1$ **then**
1. $~~~~$ $C_{00}\mapsto A_{00}\cdot B_{00}$
1. **else**
1. $~~~~$ $P_0 = \text{strassen}(A_{00},B_{01}-B_{11})$
1. $~~~~$ $P_1 = \text{strassen}(A_{00}+A_{01},B_{11})$
1. $~~~~$ $\dots$ (see Eq. ($\star$) in [this notebook](./Lecture_11a_MatrixMultiplication.ipynb#strassen_algorithm))
1. $~~~~$ $P_6 = \text{strassen}(A_{00}-A_{10},B_{00}+B_{01})$
1. $~~~~$ $C_{00} = P_4+P_3-P_1+P_5$
1. $~~~~$ $C_{01} = P_0 + P_1$
1. $~~~~$ $C_{10} = P_2 + P_3$
1. $~~~~$ $C_{11} = P_4 + P_0 - P_2 - P_6$
1. **end if**

This can be seen as a concrete example of the abstract $\text{DaQ}$ function above by observing that:

* The problem size is described by the number $n$
* The input consists of two matrices of size $n\times n$
* The matrices $A$ and $B$ are split into smaller $n/2\times n/2$ matrices $A_{00}, A_{01}, \dots, B_{00}, B_{01}, \dots$ which correspond to the chunks $S_0,S_1,\dots$ in the abstract algorithm. Hence, in our case $b=2$.
* Those chunks  are combined into $B_{01}-B_{11}, A_{00}+A_{01}, \dots$, which correspond to the chunks $S_0',S_1',\dots$
* The recursive call to the function to calculate the matrices $P_0,P_1,\dots,P_6$ (which correspond to the chunks $T_0,T_1,\dots$ in the abstract algorithm) is executed $a=7$ times.
* Finally, the $P_0,P_1,\dots,P_6$ are combined to obtain the entries of the matrix $C$

Hence, we conclude:

$$
E_{\text{matrix-matrix}}^{\text{strassen}}(n) = \begin{cases}
7E_{\text{matrix-matrix}}^{\text{strassen}}(n/2) + f(n) & \text{for $n\ge 2$}\\
E_0 & \text{otherwise}
\end{cases}
$$

In fact, in the notebook on Strassen's algorithm we worked out that $f(n)=\frac{9}{2}n^2$ and $E_0$, which results in the explicit formula for $E_{\text{matrix-matrix}}^{\text{strassen}}(n)$ written down [there](./Lecture_11a_MatrixMultiplication.ipynb#strassen_complexity).

We will now show how the computational complexity of the generic Divide-and-Conquer algorithm can be derived if we know $a$, $b$ and $f(n)$.

### Computational complexity
To work out the total cost for a general divide-and-conquer algorithm, we can draw a tree which illustrates how the algorithm is executed. For simplicity, we assume that the problem size is an integer power of $b$, i.e. $n=b^k$ with $k\in\mathbb{N}$. First, we need to execute the function for the largest value of $n$, which incurs a cost of $f(n)$. This now triggers $a$ recursive calls to the function with $n/b$. For each of those cases, we need to execute $f(n/b)$ elops and then call the function again $a$ times on a problem of size $n/b^2$. This continues until $n/b^j$ is $1$. The whole process can be visualised as an (inverted) tree; the total cost on each level of the tree is shown on the very right.

![Complexity tree](./figures/MasterTheoremTree.svg "Tree for divide and conquer algorithms")

The height $H(n)$ of the tree has to satisfy $n/b^{H(n)}=1$, so $H(n)=\log_b(n)$. The number of leaf-nodes at the very bottom is $W(n)=a^{H(n)}=a^{\log_b(n)}=n^{\log_b(a)}$. Since the problem size is $n=1$ for each of these leaf nodes, each leaf node incurs a cost of $E_0$.

The total cost is the sum over all nodes, and therefore given by

$$
E(n) = f(n) + a\cdot f(n/b) + a^2 \cdot f(n/b^2) + \dots + a^{H(n)-1}\cdot f(n/b^{H(n)-1}) + E_0 \cdot W(n)
$$

This can be written in more succinct form as:

$$
\begin{aligned}
E(n) &= E_0\cdot W(n)  + \sum_{j=0}^{H(n)-1} a^j\cdot f(n/b^j)\\
&= E_0\cdot n^{\log_b(a)} + \sum_{j=0}^{\log_b(n)-1} a^j\cdot f(n/b^j)
\end{aligned}
$$

This is quite a complicated function of $n$ which depends on $a$, $b$, $f(n)$ and $E_0$. Before we discuss how this can be simplified further with the Master Theorem, note that there are three cases. The total number of elops can be ...

* **Case 1**: ... concentrated in the leaves, i.e. most of the time is spent on dealing with the case $n=1$
* **Case 2**: ... equally distributed between all levels of the treee
* **Case 3**: ... concentrated in the root node, i.e. most of the time is spent on dealing with the case where $n$ is largest.

The following figure illustrates this, by highlighting the computationally most expensive nodes in red:

![Cases for the complexity tree](./figures/MTTreeCases.svg "Different cases for the complexity tree")

Which case we end up in depends on how fast the function $f(n)$ grows relative to the width of the tree $W(n)=n^{\log_b(a)}$.

## Master Theorem
The crucial insight is now that if we know $a$, $b$ and $f(n)$ for a specific divide-and-conquer algorithm, we can write down a function $g(n)$ such that the number of elops is $E(n)=\Theta(g(n))$.

This is expressed by the following theorem:

### Master Theorem for divide-and-conquer algorithms
Let $a\ge 1$ and $b>1$ be constants, let $f(n)$ be a function and $E(n)$ defined by

$$
E(n) = a\cdot E(n/b) + f(n) \qquad \text{for all $n\ge 2$}.
$$

Then the order of growth $E(n)$ is given as follows:

1. If $f(n)=\mathcal{O}(n^c)$ for $c<\log_b(a)$, then $E(n)=\Theta(n^{\log_b(a)})$
2. If $f(n)=\Theta(n^{\log_b(a)})$, then $E(n)=\Theta(n^{\log_b(a)}\log(n))$
3. If $f(n)=\Omega(n^{c'})$ for some constant $c'>\log_b(a)$, and if $a\cdot f(n/b) \le \gamma f(n)$ for some positive constant $\gamma<1$ and all sufficiently large $n$, then $E(n)=\Theta(f(n))$.

The proof of this theorem is beyond the scope of this course and can be found in **chapter 4.5 of "Introduction to Algorithms" by Cormen, Leiserson, Rivest and Stein**. While above we have assumed that $n$ is a power of $b$, the theorem also holds in the more general case where this is not true. Again, discussing this rigorously is beyond the scope of this course.

### Examples
Let's have a look at some examples from previous lectures and tickables:

#### [Strassen's method](../week11/Lecture_11a_MatrixMultiplication.ipynb) for matrix-matrix multiplication (week 11)
In this case we have $a=7$, $b=2$, $f(n)=\frac{9}{2}n^2=\mathcal{O}(n^2)$, and $2<\log_2(7)\approx2.80735$. Hence Case 1 of the Master Theorem applies and we have that $E(n)=\Theta(n^{\log_b(a)})=\Theta(n^{\log_2(7)})= \mathcal{O}(n^{2.80735})$. This agrees with what we found before.

#### [Merge Sort](../week10/Lecture_10_Mergesort.ipynb) (week 10)
In this case we have that $a=2$ (we need to merge two lists in each step) and $b=2$ (each of these two lists is half the size of the original list). Also, the number of operations required for merging two sorted lists is proportional to the combined length of those lists, so $f(n)=\Theta(n)=\theta(n^{\log_2(2)})$. Hence Case 2 of the Master Theorem applies, and we find that $E(n)=\Theta(n\log(n))$. Again, this is consistent with the result we previously worked out by hand.

#### The [FastPower method](../week06/Lecture_06b_DivideAndConquer.ipynb) for computing $x^n$ (week 6)
In this case $a=1$ (only one recursive call is required) and $b=2$ (we reduce the problem of computing $x^n$ to the problem of computing $(x\cdot x)^{n/2}$, so $n$ gets replaced by $n/2$ in the recursive calls). Also $f(n)=\Theta(1)=\Theta(n^{\log_2(1)})$, since we have to do a fixed number of operations for each $n$. Hence Case 2 of the Master Theorem applies and we have that $E(n)=\Theta(\log(n))$.

#### [Binary Search](../week07/Tickable_07.ipynb) (Tickable 7)
Again, we have that $a=1$ and $b=2$ (we reduce the problem of finding a number $x$ in a list of $n$ elements to the problem of finding $x$ in a list of half that size). In each call to the $\text{binary_search()}$ function we make one comparison ($x$ is compared to $a_{n^*}$). Hence again Case 2 of the Master Theorem applies and we have for the number of comparisons $C(n)=\Theta(\log(n))$.

#### Ficticious example
Assume that $a=3$, $b=2$ and $f(n)=n^2$. We have that $2>\log_2(3)\approx 1.5850$  and also that $3\cdot f(n/2)=\frac{3}{4}n^2\le \gamma f(n)$ with $\gamma=\frac{3}{4}$. Hence Case 3 of the Master Theorem applies and $E(n)=\Theta(n^2)$.

## Check your understanding

### Exercise 1
Consider a divide-and-conquer algorithm with $a=b>0$ and $f(n)=2\sqrt{n}+3$. Using the Master Theorem, what is the computational complexity?

1. $\Theta(\sqrt{n})$
2. $\Theta(n)$
3. $\Theta(n\log(n))$
4. Can not be deduced with the Master Theorem

Answer 2 is correct. The function $f(n)$ is $\mathcal{O}(\sqrt{n})=\mathcal{O}(n^c)$ with $c=1/2$. Since $a=b$ we have that $1/2=c<\log_b(a)=1$ and case 1 of the Master Theorem applies. The computational complexity is $E(n)=\Theta(n^{\log_b(a)})=\Theta(n)$.

### Exercise 2
Consider a divide-and-conquer algorithm with $a=8$, $b=2$ and $f(n)=7n^3+3n+1$. Using the Master Theorem, what is the computational complexity?

1. $\Theta(n^3)$~
2. $\Theta(n\log(n))$
3. $\Theta(n^3\log(n))$
4. Can not be deduced with the Master Theorem

Answer 3 is correct. In this case $f(n)=\Theta(n^3)=\Theta(n^{\log_2(8)}=\Theta(n^{\log_b(a)}$ and Case 2 of the Master Theorem applies: the computational complexity is $\Theta(n^3\log(n))$. 

### Exercise 3
Consider a divide-and-conquer algorithm with $a=9$, $b=3$ and $f(n)=(2\sqrt{n}+3)n^2$. Using the Master Theorem, what is the computational complexity?

1. $\Theta(n^{5/2})$
2. $\Theta(n^2)$
3. $\Theta(n^{2}\log(n))$
4. Can not be deduced with the Master Theorem

Answer 1 is correct. So see this, not that $f(n)=\Omega(n^2\sqrt{n})=\Omega(n^c)$ with $c'=5/2$. But $5/2=c'>\log_b(a)=\log_3(9)=2$ so we need to consider Case 3 of the Master Theorem. To verify the condition $a\cdot f(n/b)\le \gamma f(n)$ for some $\gamma<1$ we pick some $\frac{1}{\sqrt{3}}<\gamma<1$. Some straightforward algebra shows that for all $n>\frac{27}{4(\sqrt{3}\gamma-1)^2}$ we have that $\frac{2}{\sqrt{3}}\sqrt{n}+3<2\gamma \sqrt{n}$. This implies that

$$
a\cdot f(n/b) = 9\cdot \left(2\sqrt{n/3}+3\right)\left(\frac{n}{3}\right)^2 = \left(\frac{2}{\sqrt{3}}\sqrt{n}+3\right)n^2 < 2\gamma n^2 \sqrt{n} < \gamma (2\sqrt{n}+3)n^2 = \gamma f(n)
$$

Hence, Case 3 of the Master Theorem indeed applies and the computational complexity is $\Theta(f(n)) = \Theta(n^{5/2})$.

### Exercise 4
Consider a divide-and-conquer algorithm with $a=1$, $b=2$ and $f(n)=n(2-(-1)^n)$. Using the Master Theorem, what is the computational complexity?

1. $\Theta(n)$
2. $\Theta(\sqrt{n})$
3. $\Theta(1)$
4. Can not be deduced with the Master Theorem

In this particular example the computational complexity cannot be deduced with the Master Theorem.

Observe that $\log_b(a)=\log_2(1)=0$. The function $f(n)$ is $\mathcal{O}(n)=\mathcal{O}(n^c)$ with $c=1$ and $\Omega(1)=\Omega(n^{c'})$ with $c'=0$. Case 1 of the Master Theorem does not apply since $1=c>\log_b(a)=0$. Case 3 does not apply either since $0=c'\le \log_b(a)=0$. Since $f(n)$ is not $\Theta(n)$, we can also rule out Case 2.

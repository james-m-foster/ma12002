## Fibonacci numbers
The [Fibonacci numbers](https://en.wikipedia.org/wiki/Fibonacci_number) are defined as

$$
F_0 = 0,\qquad
F_1 = 1,\qquad
F_n = F_{n-1}+F_{n-2}\quad \text{for $n\ge 2$},
$$

For example, the next four Fibonacci numbers are:

$$
\begin{aligned}
F_2&=F_1+F_0=0+1=1,\\
F_3&=F_2+F_1=1+1=2,\\
F_4&=F_3+F_2=1+2=3,\\
F_5&=F_4+F_3=2+3=5,\\
\end{aligned}
$$

Pseudocode for calculating the $n$-th Fibonacci number is given in the following algorithm. Recall that in pseudocode $a\mapsto 1$ means: *''Assign the variable $a$ to $1$''* Similarly, $a\mapsto x$ means *''Assign the variable $a$ to the value stored in the variable $x$''*

**Input** $n$
**Output** $n$-th Fibonacci number

1. $~~~~$ **return** $1$
1. **else**
1. $~~~~$ Initialise $\alpha\mapsto 0$; $\beta\mapsto 1$
1. $~~~~$ **for** $k=1,\dots,n-1$

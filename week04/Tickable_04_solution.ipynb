## &#9745; Task 1
The *iterative* implementation is given by the following Python code:

```python
def fib_iter(n):
    '''Compute n-th Fibonacci number iteratively

    The Fibonacci numbers are defined as:
    F_0 = 0, F_1 = 1, F_n = F_{n-1}+ F_{n-2} for n >= 2

    input: Number n
    output: n-th Fibonacci number F_n
    '''
    
    if not (n == int(n)):
        raise Exception('Fibonacci number F_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Fibonacci number F_n not defined for negative values')
    elif n==0:
        return 0
    elif n==1:
        return 1
    else:
        alpha = 0; beta = 1
        for k in range(1,n):
            gamma = beta + alpha
            alpha = beta
            beta = gamma
        return gamma
```

The order of the assignments in lines 8-10 of the pseudocode matter, and the result will change if any of these lines are swapped.

## &#9745; Task 2
The *recursive* implementation is given by the following Python code:

```python
def fib_rec(n):
    '''Compute n-th Fibonacci number recursively.

    The Fibonacci numbers are defined as:
    F_0 = 0, F_1 = 1, F_n = F_{n-1}+ F_{n-2} for n >= 2
    
    input: Number n
    output: n-th Fibonacci number F_n
    '''
    if not (n == int(n)):
        raise Exception('Fibonacci number F_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Fibonacci number F_n not defined for negative values')
    elif n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fib_rec(n-1)+fib_rec(n-2)
```

It is important that the we use `fib_rec()` (and not `fib_iter()`) in the recursive call in the final line. We can check that the implementations pass the provided tests.

## Further ideas
1. First of all, $F_1$ is only computed once with `fib_iter()` (by assigning $\beta \mapsto 1$ in line 6. of the pseudocode). To work out how often `fib_rec(1)` is called in the recursive implementation, consider the following graph, where each arrow corresponds to a recursive call:

![Callgraph for fib_rec(4)](./figures/fibonacci_callgraph.svg "Callgraph for fib_rec(4)")

For example, three boxes in the first two lines of the above diagram say that when we call `fib_rec(4)`, this will recusively call `fib_rec(3)` and `fib_rec(2)`.

2. Let $\nu_n$ be the number of times `fib_rec(1)` is called when computing the $n$-th Fibonacci number $F_n$. Then obviously $\nu_1=1$ and $\nu_0=0$. But we also have that $\nu_n=\nu_{n-1}+\nu_{n-2}$ for $n\ge 2$, since `fib_rec(n)` will recursively call `fib_rec(n-1)` and `fib_rec(n-2)`, so the number of calls to `fib-rec(1)` in `fib_rec(n)` has to be the sum of the number of calls to `fib_rec(1)` in `fib_rec(n-1)` and `fib_rec(n-2)`. Hence $\nu_n=F_n$, i.e. when computing the $n$-th Fibonacci number with `fib_rec()` we call `fib_rec(1)` exactly $F_n$ times.

We will look at this in more detail later in the course, but for now let us imagine that it takes some time $\delta t>0$ to execute `fib_iter(1)` (this is reasonable, since (as shown in point 2.) we have to execute at least the if-statement in that case). Since we call `fib_iter(1)` $F_n$ times when computing the $n$-th Fibonacci number with `fib_rec(n)`, the time it takes to execute `fib_rec(n)` is at least $t_{\min}(n) = F_n\cdot \delta t$. With the lower bound on $F_n \ge \left(\frac{3}{2}\right)^{n-1}$ given in the Tickable we find that
$$
t_{\min}(n) = F_n\cdot \delta t \ge \left(\frac{3}{2}\right)^{n-1}\cdot \delta t.
$$
But $\left(\frac{3}{2}\right)^{n-1}$ is a function which grows rapidly, and this explains the strong increase in the runtime for `fib_rec(n)`: if we increase $n$ by $1$ the runtime goes up by a factor of at least $1.5$.

5. Iterative and recursive Python functions for computing the $n$-th Tribonacci number are given as follows:

```python
def trib_iter(n):
    '''Iteratively compute the n-the Tribonacci number T_n where 
    T_0=T_1=0, T_2=1 and T_n = T_{n-1}+T_{n-2}+T_{n-3} for n >= 3
     
    input
      number n
    returns
      n-th Tribonacci number T(n)
    '''
    if not (n == int(n)):
        raise Exception('Tribonacci number T_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Tribonacci number T_n not defined for negative values')
    elif (n==0) or (n==1):
        return 0
    elif n==2:
        return 1
    else:
        alpha = 0; beta = 0; gamma = 1
        for k in range(2,n):
            delta = gamma + beta + alpha
            alpha = beta
            beta = gamma
            gamma = delta
        return delta

def trib_rec(n):
    '''Recursively compute the n-the Tribonacci number T_n where 
    T_0=T_1=0, T_2=1 and T_n = T_{n-1}+T_{n-2}+T_{n-3} for n >= 3
     
    input
      number n
    returns
      n-th Tribonacci number T(n)
    '''
    if not (n == int(n)):
        raise Exception('Tribonacci number T_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Tribonacci number T_n not defined for negative values')
    elif (n==0) or (n==1):
        return 0
    elif n==2:
        return 1
    else:
        return trib_rec(n-1)+trib_rec(n-2)+trib_rec(n-3)
```

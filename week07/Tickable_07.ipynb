## Binary Search
Imagine that we are given a sorted list of $n$ distinct integer numbers $a = [a_0, a_1,\dots, a_{n-1}]$ with $a < a_1 < \dots < a_{n-1}$ and want to check whether this list contains a particular number $x$. This number $x$ is often called the ''target''. Expressing this task more mathematically, we are looking for the index $j \in [0,n-1]$ such that $a_j = x$. If the target is not contained in the list $a$, we set $j\mapsto \infty$. The time $T(n)$ it takes to execute this algorithm for some given $n$ can be quantified by counting the number of times we compare $x$ to some number $a_i$ from the list, i.e.

$$
T(n) = \alpha \cdot C(n)\qquad\text{with $C(n)=$ number of comparisons of $x$ with some $a_i$}
$$

Here $\alpha$ is some constant that we are not interested in since we are really concerned with the relative growth of the runtime as $n$ increases (see the discussion in the [lecture on orders of growth](Lecture_07_OrderOfGrowth.ipynb)). One way of finding $x$ is to inspect the elements in the list one by one, comparing each entry to the target $x$. This leads to the following brute-force algorithm:

### Algorithm: brute force search
**Input**: Sorted list $a = [a_0, a_1,\dots, a_{n-1}]$, integer $x$ (the target)
**Output**: Index $j$ such that $a_j=x$ (or $\infty$ if $x\not\in a$)
1. Set $j\mapsto\infty$
1. **for** $k=0,1,\dots,n-1$
1. $~~~~$ **if** $a_k==x$ **then**
1. $~~~~~~~~$ Set $j\mapsto k$
1. $~~~~$ **end if**
1. **end for**

The corresponding Python code is this ($\infty$ is represented by the numpy constant `np.infty`):

````python
import numpy as np
def brute_force_search(a,x):
    '''Find x in list a=[a_0,a_1,...,a_{n-1}] with a brute force method
    
    input:
      * list a=[a_0,a_1,...,a_{n-1}] of length n
      * number x to be found
    returns:
      index j such that a_j = x or infinity if the list does not contain x
    '''
    n = len(a)
    j = np.infty
    for k in range(n):
        if a[k]==x:
            j = k
    return j
````

#### Divide and conquer approach
However, we can do better than this. Recall that for any real number $x$ the floor of $x$, denoted by $\lfloor x\rfloor=\max_{m\in\mathbb{Z}}\{m\le x\}$ is the largest integer that is less than or equal to $x$. For example

$$
\lfloor 1.1\rfloor = 1,\quad \lfloor -1.1\rfloor=-2,\quad\lfloor 4\rfloor  = 4,\quad\lfloor 3.999\rfloor = 3.
$$

For the special case $x=\frac{n}{2}$ we find that

$$
\lfloor \frac{n}{2}\rfloor = \max_{m\in\mathbb{Z}}\left\{m\le \frac{n}{2}\right\}=
\begin{cases}
\frac{n}{2} & \text{for $n$ even,}\\
\frac{n-1}{2} & \text{for $n$ odd.}
\end{cases}
$$

To simplify notation, let us denote this number by $n^*=\lfloor\frac{n}{2}\rfloor$ from now on. We start by comparing $x$ to $a_{n^*}$, the element in the "middle" of our array. Since the list is sorted, if $x$ is smaller than $a_{n^*}$, then $x$ must be contained in the list $a^{(\text{left})}=[a_0,a_1,\dots,a_{n^*-1}]$. Otherwise, $x$ must be contained in the list $a^{(\text{right})}=[a_{n^*},a_{n^*+1},\dots,a_{n-1}]$. Next, we can apply the same strategy to either $a^{(\text{left})}$ or $a^{(\text{right})}$. Note in particular that the number of elements in each of the lists $a^{(\text{left})}$ and $a^{(\text{right})}$ is no larger than $\frac{n}{2}$, so we reduced the problem to a smaller problem of (no more than) half the size.

This strategy can be applied recursively until we arrive at a list which is either empty or contains only one element. In the latter case we compare this single element to the target. We arrive at the following Divide and Conquer method:

### Algorithm $\boldsymbol{\textbf{binary_search}(a,x)}$
**Input**: Sorted list $a = [a_0, a_1,\dots, a_{n-1}]$, integer number $x$
**Output**: Index $j$ such that $a_j=x$ (or $\infty$ if $x\not\in a$)
1. **if** $a$ is empty list **then**
1. $~~~~$ **return** $\infty$
1. **else if** $n==1$ **then**
1. $~~~~$ **if** $a_0==x$ **then**
1. $~~~~~~~~$ **return** $0$
1. $~~~~$ **else**
1. $~~~~~~~~$ **return** $\infty$
1. $~~~~$ **end if**
1. **else**
1. $~~~~$ Set $n^*=\lfloor \frac{n}{2}\rfloor$
1. $~~~~$ **if** $x<a_{n^*}$ **then**
1. $~~~~~~~~$ Set $a^* \mapsto a^{(\text{left})}= [a_0,a_1,\dots,a_{n^*-1}]$
1. $~~~~~~~~$ **return** $\text{binary_search}(a^*,x)$
1. $~~~~$ **else**
1. $~~~~~~~~$ Set $a^* \mapsto a^{(\text{right})}=[a_{n^*},a_{n^*+1},\dots,a_{n-1}]$
1. $~~~~~~~~$ **return** $n^* + \text{binary_search}(a^*,x)$
1. $~~~~$ **end if**
1. **end if**

### &#9745; Task 2
In Python, `binary_search()` can be implemented recursively as follows:

```python
import numpy as np

def binary_search(a,x):
    '''Find x in list a=[a_0,a_1,...,a_{n-1}] with a the binary search method
    
    input:
      * list a=[a_0,a_1,...,a_{n-1}] of length n
      * number x to be found
    returns:
      index j such that a_j = x or infinity if the list does not contain x
    '''
    n = len(a)
    if n==0:
        return np.inf
    elif n==1:
        if a[0] == x:
            return 0
        else:
            return np.inf
    else:
        n_star = n//2
        if (x < a[n_star]):
            a_star = a[:n_star]
            return binary_search(a_star,x)
        else:
            a_star = a[n_star:]
            return n_star + binary_search(a_star,x)
```

## Further things to consider
#### Number of recursive calls
Let us denote the number of recursive calls if binary search for a list of length $n$ with $\nu(n)$. Observe that In each recursive call the function calls itself for a list of length $n^*=\lfloor\frac{n}{2}\rfloor$ and thus we have the recursion

$$
\nu(n) = 1 + \nu\left(\lfloor\frac{n}{2}\rfloor\right)\qquad\text{with $\nu(1)=\nu(0)=0$.}
$$

Exactly the same recursion relation held for the fast power method (see the detailed derivation at the end of the the [lecture on Divide and Conquer](../week06/Lecture_06b_DivideAndConquer.ipynb)). There we found that 

$$
\log_2(n) -1 < \nu(n) \le \log_2(n)
$$

and hence $\nu(n)=\Theta(\log_2(n))$. If we assume that each recursive call involves a fixed amount out work (for example, to execute the if-statements), then it seems natural that the runtime will also be $\Theta(\log_2(n))$.

#### Duplicate elements
If the list contains duplicate elements (but is still sorted in the sense that $a_j\le a_k$ for $j<k$), the binary search algorithm will always return the index of the rightmost element that is equal to the target. Consider for example:

```python
a = [1,3,5,5,5,6,8,12]
x = 5
binary_search(a,x)
```

#### Unsorted list
If the list is unsorted and we sort it with mergesort before applying binary search, we expect the runtime to be

$$
T_1(n) = \Theta(n\log_2(n)) + \Theta(\log_2(n)) = \Theta(n\log_2(n))\qquad(\dagger)
$$

This essentially means that for very large values of $n$ the cost will be dominated by the cost of sorting the list.

It might be a bit odd that we can add the ''$\Theta$'s'' for two functions in Eq. ($\dagger$). This addition can be defined mathematically, but instead it is easier to remember that if $f(n)=\Theta(g(n))$ this essentially means that for $n\gg 1$ we can approximate $f(n)\approx C\cdot g(n)$ with some constant $C$. Hence the equation above says that for very large values of $n$ we have that

$$
T_1(n) \approx C_1\cdot n\log_2(n) + C_2\cdot \log_2(n) = C_1\cdot\left(1+\frac{C_2}{n}\right)n\log_2(n) 
\approx C_1\cdot n\log_2(n)\qquad\text{for $n\gg 1$}.
$$

On the other hand, if we simply apply the brute force search algorithm this will have a cost of

$$
T_2(n) = \Theta(n)
$$

since we have to compare the target to $n$ elements, and it is sensible to assume that there is a fixed cost for each individual comparison. For large values of $n$ this is more efficient since the function $n\log_2(n)$ grows faster than $n$.

### Challenge
There are different ways of implementing binary search non-recursively. However, the key idea is to introduce two indices $i_{\text{low}}$, $i_{\text{high}}$ and update these indices such that the target can only be contained in the list $[a_{i_{\text{low}}},a_{i_{\text{low}}+1},\dots,a_{i_{\text{high}}-1},a_{i_{\text{high}}}]$. If $i_{\text{low}}>i_{\text{high}}$ then the target is not contained in the array.

In each step, we first compute $i_{\text{middle}} = \lfloor \frac{i_{\text{low}}+i_{\text{high}}}{2} \rfloor$. If $x<a_{i_\text{middle}}$ then the target can only be contained in $[a_{i_{\text{low}}},a_{i_{\text{low}}+1},\dots,a_{i_{\text{middle}}-2},a_{i_{\text{middle}}-1}]$, so we set $i_{\text{high}}\mapsto i_{\text{middle}}-1$ and leave $i_{\text{low}}$ as it is. If $x>a_{i_\text{middle}}$, then the target can only be contained in $[a_{i_{\text{middle}}+1},a_{i_{\text{middle}}+2},\dots,a_{i_{\text{high}}-1},a_{i_{\text{high}}}]$, so we set $i_{\text{low}}\mapsto i_{\text{middle}}+1$ and leave $i_{\text{high}}$ unchanged. Finally, if $x=a_{i_\text{middle}}$ we have found the target and return the index $i_{\text{middle}}$. We keep repeating this as long as $i_{\text{low}}\le i_{\text{high}}$. Note that in each step the difference $i_{\text{high}}-i_{\text{low}}$ decreases by at last one, so the iteration will eventually stop. If this is the case, we conclude that the target is not contained in the list and return $\infty$.

This algorithm can be implemented as a Python function `binary_search_iter()` as follows:

```python
def binary_search_iter(a,x):
    '''Find x in list a=[a_0,a_1,...,a_{n-1}] with a the binary search method
    
    input:
      * list a=[a_0,a_1,...,a_{n-1}] of length n
      * number x to be found
    returns:
      index j such that a_j = x or infinity if the list does not contain x
    '''
    n = len(a)
    if n==0:
        return np.inf
    else:
        i_low = 0
        i_high = n - 1
        
        while i_low <= i_high:
            i_middle = (i_high + i_low)//2
            if x < a[i_middle]:
                i_high = i_middle - 1
            elif a[i_middle] < x:
                i_low = i_middle + 1
            else:
                return i_middle
            
        return np.inf
```

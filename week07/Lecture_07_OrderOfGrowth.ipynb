# Orders of growth
In one of the previous lectures we started to analyse the performance of two algorithms for evaluating polynomials by counting elementary operations (elops). In this notebook we will develop the mathematical tools to quantify and classify this growth further. For this we will use so-called big-O (pronounced ''big-Oh'') notation.

## Polynomial evaluation revisited
Recall that we counted the number of elops for two algorithms for evaluating degree-$n$ polynomials of the form

$$
p(x) = a_0 + a_1 x + a_2 x^2 + \dots + a_n x^n = \sum_{j=0}^n a_j x^j
$$

1. The ***naive method***, which computes each power $x^j$ with $j-1$ multiplications and then adds the individual polynomials $a_jx^j$.

2. ***Horner's rule*** which brackets the polynomial in a clever way and reduces the evaluation to a sequence of multiplications and additions.

The corresponding numbers of elops as a function of the polynomial degree $n$ turned out to be:

$$
\begin{aligned}
E_{\text{naive}}(n) &= \frac{1}{2}n^2 + \frac{3}{2}n,\\
E_{\text{Horner}}(n) &= 2n.
\end{aligned}
$$

We saw that for large $n$ Horner's rule is much more efficient. The reason for this is that the highest power of $n$ in $E_{\text{Horner}}(n)$ is one, but in the expression for $E_{\text{naive}}(n)$ the highest power of $n$ is two. In other words, for large values of $n$ the number of elops for Horner's rule grows linearly, whereas for the naive scheme it increases approximately quadratically with $n$. More explicitly, we observe that:

* For large values of $n$, the term $\frac{3}{2}n$ in $E_{\text{naive}}(n)$ has very little impact, its relative importance for the runtime decreases in proportion to $1/n$.
* While the coefficients of the highest order terms ($2n$ for Horner's rule and $\frac{1}{2}n^2$ for the naive method) influence the *absolute* runtime, they do not incluence the *relative growth*. This relative growth can for example be quantified by asking by which factor the runtime increases if $n$ is doubled.

We conclude that for practical purposes the exact expressions for the number of elops $E(n)$ contain a lot of irrelevant information. All that really matters in this example is the exponent of the highest poiwer of $n$. To highlight this, here are the expressions for the runtime again, with the important terms in red and all irrelevant information grayed out:

$$
\begin{aligned}
E_{\text{naive}}(n) &= \color{gray}{\frac{1}{2}}\color{red}{n^2} \color{gray}{+ \frac{3}{2}n},\\
E_{\text{Horner}}(n) &= \color{gray}{2}\color{red}{n}.
\end{aligned}
$$


In the following we will develop the mathematical technology for extracting those important terms for general functions $E(n)$. For this, we first discuss the asymptotic growth of arbitrary functions, which can be quantified with so-called big-O notation.

## Big-O notation
We start with some definitions. In the following we always assume that the functions $f$ and $g$ are defined for all non-negative integer numbers $n\in\mathbb{N}_0$. We further assume that $g$ is non-negative valued, i.e. $g(n)\ge 0$ for all $n\in\mathbb{N}_0$ and that there exists some integer $n_+$ such that $f(n)\ge 0$ for all $n\in\mathbb{N}_0$ with $n\ge n_+$.

### Definition (big-O):
Let $g$ be a non-negative-valued function. The function $f$ is of order $\mathcal{O}(g)$ (pronounced *''order big-O of $g$''*) if there exists a positive constant $c$ and a positive integer $n_0$ such that

$$
f(n) \le c\cdot g(n)\qquad\text{for all $n\ge n_0$}
$$

### Definition (big-Omega):
Let $g$ be a non-negative-valued function. The function $f$ is of order $\Omega(g)$ (pronounced *''order big-Omega of $g$''*) if there exists a positive constant $c$ and a positive integer $n_0$ such that

$$
f(n) \ge c\cdot g(n)\qquad\text{for all $n\ge n_0$}
$$

### Definition (big-Theta):
Let $g$ be a non-negative-valued function. The function $f$ is of order $\Theta(g)$ (pronounced *''order big-Theta of $g$''*) if there exist positive constants $c_1$, $c_2$ and a positive integer $n_0$ such that

$$
c_1 \cdot g(n) \le f(n) \le c_2\cdot g(n)\qquad\text{for all $n\ge n_0$}
$$

The following figures illustrate this:

![Graphical illustration of different orders of growth](./figures/ordersofgrowth.svg "Orders of growth: $\mathcal{O}$ (left), $\Omega$ (centre) and $\Theta$ (right).")

To make the notation more concise, we usually write $f(n) = \mathcal{O(g(n))}$ if $f$ is of order $\mathcal{O}(g)$ and so on.

One can show that following lemma holds:

### Lemma
Let $f$ and $g$ be integer valued functions. Then $f(n)=\Theta(g(n))$ if and only if $f(n)=\mathcal{O}(g(n))$ and $f(n)=\Omega(g(n))$.

You should convince yourself that you can show why this is true - the proof only requires the above definitions and basic logic.

### Transitivity
Using the definitions, we can make a few observations:

* $g_1(n)=\mathcal{O}(g_2(n))$ and $g_2(n) = \mathcal{O}(g_3(n))$ imply that $g_1(n) = \mathcal{O}(g_3(n))$
* $g_1(n)=\Omega(g_2(n))$ and $g_2(n) = \Omega(g_3(n))$ imply that $g_1(n) = \Omega(g_3(n))$
* $g_1(n)=\Theta(g_2(n))$ and $g_2(n) = \Theta(g_3(n))$ imply that $g_1(n) = \Theta(g_3(n))$

The first two statements are straightforward to prove by using the definitions of $\mathcal{O}$, $\Omega$ and $\Theta$ above and the third statementment immediately follows from the lemma. Mathematically, the three statements statements express the fact that the relations $f(n)=\mathcal{O}(g(n))$, $f(n)=\Omega(g(n))$ and $f(n)=\Theta(g(n))$ are *transitive*. You might want to compare this to the relations ''$\le$'', ''$\ge$'' and ''$=$'' between real numbers. For $a,b,c \in\mathbb{R}$ we have that:

* $a\le b$ and $b\le c$ imply that $a\le c$
* $a\ge b$ and $b\ge c$ imply that $a\ge c$
* $a=b$ and $b=c$ imply that $a=c$

Just as ''$\le$'' and ''$\ge$'' order real numbers, you can think of big-O notation as a way of ordering functions.

### Examples
Let us consider a few examples to illustrate the ideas behind big-O notation:

* ***Example 1***: Let $$f(n)=\frac{1}{3}n^3+2n^2+4.$$ Set $g(n)=n^3$ and consider 

$$g(n)-f(n)=\frac{2}{3}n^3 \left(1-\left[\frac{3}{n}+\frac{6}{n^3}\right]\right)$$

The term in square brackets decreases with $n$. It is smaller than $1$ if $n\ge 6$ and hence $g(n)\ge f(n)$ for $n\ge 6$. Setting $c=1$ and $n_0=6$, this implies that $f(n)=\frac{1}{3}n^3+2n^2+4\le n^3 = c\cdot n^3=c\cdot g(n)$ for $n\ge n_0$ and hence $\frac{1}{3}n^3+2n^2+4=\mathcal{O}(n^3)$. We also find that  $\frac{1}{3}n^3+2n^2+4 \ge \frac{1}{3} n^3$ for all $n\ge 1$, and hence $\frac{1}{3}n^3+2n^2+4 = \Omega(n^3)$ with $c=\frac{1}{3}$, $n_0=1$. Together this implies that

$$
\frac{1}{3}n^3+2n^2+4 = \Theta(n^3)
$$

* ***Example 2***: Let $$f(n)=\frac{n^3+2n+1}{n+2}$$ and consider $g(n)=n^2$. Then

$$
g(n)-f(n) = n^2 - \frac{n^3+2n+1}{n+2} = \frac{2n(n-1)-1}{n+2}
$$

If $n\ge 2$ this difference is positive, and hence $f(n)=\mathcal{O}(n^2)$ with $c=1$, $n_0=2$. Further,

$$
f(n)-\frac{1}{2}g(n) = \frac{n^3+2n+1}{n+2}-\frac{1}{2}n^2 = \frac{n^3-2n^2+4n+2}{2(n+2)} = \frac{n^3}{2(n+2)}\left(1-\left[\frac{2}{n}-\frac{4}{n^2}-\frac{2}{n^3}\right] \right)
$$

For $n\ge 4$ the term in square brackets is smaller than $1$, and hence $f(n)\ge \frac{1}{2}g(n)$ for $n\ge 4$. This implies that $f(n)=\Omega(g(n))$ with $c=\frac{1}{2}$, $n_0=4$. We finally conclude that $f(n)=\Theta(n^2)$.

* ***Example 3***: Let

$$
f(n) = \begin{cases}
n^7 & \text{for even $n$}\\
1 & \text{for odd $n$}
\end{cases}
$$

Since $1\le n^7$ for $n\ge 1$ it is obvious that $f(n)=\mathcal{O}(n^7)$. However, it it not true that $f(n)=\Theta(n^7)$ since this would imply that there is a *positive* constant $c$ *independent of $n$* such that $c\cdot n^7 \le f(n)$ for $n$ sufficiently large. This constant can not exist, since it would imply that $c n^7\le 1$ for all even numbers $n$.

* ***Example 4***: Let

$$
f(n) = 3\log_2(n) + \sqrt{n}
$$

First write

$$
f(n) = \sqrt{n}\left(1+3 \frac{\log_2(n)}{\sqrt{n}}\right)
$$

and observe that $\log_2(n) \le \sqrt{n}$ for $n\ge 16$ (you can see this for example by showing that the function $\phi(x)=\frac{\log_2(x)}{\sqrt{x}}$ is decreasing since $\phi'(x)<0$ for $x\ge e^2$ and satisfies $\phi(16)=1$). This implies that $f(n)\le 4 \sqrt{n}$ for $n\ge 16$ and hence $f(n) = 3\log_2(n) + \sqrt{n}=\mathcal{O}(\sqrt{n})$ with $c=4$ and $n_0=16$.

* ***Example 5***: Consider the factorial function for $n\ge 1$

$$
f(n) = n! = n\cdot (n-1)\cdot (n-2) \cdots 3 \cdot 2 \cdot 1
$$

Since each of the $n$ factors in this expression is smaller or equal to $n$, we immediately find that $n!\le n^n$ and hence $n!=\mathcal{O}(n^n)$.

### Some general remarks
We are usually interested in the tightest upper bound on the growth of a function, i.e. we want to find the ''best'' $g$ such that $f(n)=\mathcal{O}(g(n))$. Coming back to the number of elops $E_{\text{naive}}(n)$ required for the naive polynomial evaluation algorithm, we know that $E_{\text{naive}}(n)=\mathcal{O}(n^2)$. It is also trivially true that $E_{\text{naive}}(n)=\mathcal{O}(n^3)$ and more generally $E_{\text{naive}}(n)=\mathcal{O}(n^q)$ for all $q\ge 2$. However, it is *not* true that $E_{\text{naive}}(n)=\Theta(n^q)$ for $q>2$. In other words, $E_{\text{naive}}(n)=\mathcal{O}(n^2)$ is the tightest possible upper bound for the computational complexity of the naive polynomial evaluation algorithm.

Example 3 above also demonstrates that in some cases it is not possible to find an increasing function $g$ such that $E(n)=\Theta(g(n))$.

Often the number of elops in an algorithm can be written as a polynomial in the problem size $n$. In this case we the asymptotic growth of the runtime is given by the highest power of $n$, as stated by the following theorem:

#### Theorem (complexity of polynomial functions)
Let $$f(n)=b_0 + b_1 n + b_2 n^2 + \dots + b_p n^p$$ be a polynomial of degree $p$ with $b_p>0$. Then $$f(n) = \Theta(n^p)\quad\text{(which obviously implies that $f(n)=\mathcal{O}(n^p)$ and $f(n)=\Omega(n^p)\;).$}$$

A (non-examinable) proof of this theorem can be found at the end of this notebook.

## Orders of growth of polynomial evaluation algorithms
With this knowledge we can now go back to the elop counts for out two methods for polynomial evaluation, and we find that:

$$
\begin{aligned}
E_{\text{naive}}(n) &= \frac{1}{2}n^2 + \frac{3}{2} n = \Theta(n^2)\\
E_{\text{Horner}}(n) &= 2n = \Theta(n)
\end{aligned}
$$

Observe that this big-O notation contains all the information we are really interested in: for the naive method, the cost grows by no more than a factor of four as $n$ doubles, and for Horner's rule the number of elops increases in proportion to the polynomial degree $n$.

### Using fast_power()
What about using the naive method for polynomial evaluation, but computing the powers of $x$ with the $\text{fast_power}()$ method from last week's lecture? Recall that with this method the number of elops required to compute $x^k$ is no larger than $2\log_2 (k)$. If we write the polynomial as

$$
p(x) = a_0 + a_1\cdot x + a_2 \cdot x^2 + \dots + a_n \cdot x^n = a_0 + \sum_{k=1}^{n} a_k\cdot x^k
$$

we see that we need $n-1$ additions, $n-1$ multiplications (to compute the products of $x^k$ with the coefficients $a_k$) plus the elops required for computing $x^2, x^3, x^4, \dots, x^n$. Hence, the total number of elops can be bounded by

$$
E_{\text{naive_fast_power}}(n) \le 2\left(n-1 + \sum_{k=1}^{n} \log_2(k) \right)\qquad{(\dagger)}
$$


This is quite a complicated expression, but remember that we are really only interested in the behaviour for large values of $n$. In other words, we want to find a function $g$ such that $E_{\text{naive_fast_power}}(n)=\mathcal{O}(g(n))$. To bound $E_{\text{naive_fast_power}}(n)$ for large values of $n$, first recall from Example 5 above that $n!\le n^n$ Taking the logarithm of this inequality we find that

$$
\sum_{k=1}^{n}\log_2 n \le n\log_2 n
$$

and hence we can bound the sum in Eq. $(\dagger)$ above by
$$
\begin{aligned}
E_{\text{naive_fast_power}}(n) &\le 2\left(n-1 + \sum_{k=1}^{n} \log_2(k) \right)\\
&\le 2\left(n-1 + n\log_2(n)\right)\\
&= 4n\log_2(n) + 2\underbrace{\left(n-1 - n\log_2(n)\right)}_{\text{$<0$ for $n\ge 2$}}\\
&\le 4n\log_2(n) \qquad\text{for $n\ge 2$}
\end{aligned}
$$

We conclude that $E_{\text{naive_fast_power}}(n) = \mathcal{O}(n\log_2(n))$ with $c=4$ and $n_0=2$. It can also be shown (see the end of this notebook) that $E_{\text{naive_fast_power}}(n) = \Omega(n\log_2(n))$ and hence $E_{\text{naive_fast_power}}(n)=\Theta(n\log_2(n))$. We conclude that although using the $\text{fast_power}$ method to evaluate $x^k$ in the naive evaluation of the polynomial makes the method asymptotically faster, for large values of $n$ it is still not quite as fast as Horner' s rule which is $\Theta(n)$.

## Runtime comparisons

Finally, let us actually measure the runtimes of the different methods. We will do this for large values of $n$ to check that the asymptotic expressions derived above are indeed correct. Here are the Python implementations again (make sure you execute those cells first):

````python
# import model solutions module to be able to use the fast_power() method
import model_solutions

def poly_naive(a_coeff,x):
    '''Evaluate polynomial p(x) = sum_{j=0}^{n} a_j*x^j with the naive method
    
    input:
      * a_coeff: List with polynomial cofficients a_0,a_1, ..., a_n
      * x: value at which to evaluate the polynomial
    output:
      value of polynomial p(x)
    '''
    degree = len(a_coeff)-1     # The list a_coeff contains n+1 elements, where n is the degree 
    poly = a_coeff[0]    
    for j in range(1,degree+1):   # Loop from 1 to n
        aj_xj = a_coeff[j]
        for k in range(j):
            aj_xj *= x
        poly += aj_xj # Update polynomial
    return poly

def poly_horner(a_coeff,x):
    '''Evaluate polynomial p(x) = sum_{j=0}^{n} a_j*x^j with Horner's rule
    
    input:
      * a_coeff: List with polynomial cofficients a_0,a_1, ..., a_n
      * x: value at which to evaluate the polynomial
    output:
      value of polynomial p(x)
    '''
    degree = len(a_coeff)-1     # The list a_coeff contains n+1 elements, where n is the degree 
    poly = a_coeff[degree]    
    for j in range(degree-1,-1,-1):   # Loop from 1 to n
        poly = a_coeff[j] + x*poly # Update polynomial
    return poly

def poly_naive_fast_power(a_coeff,x):
    '''Evaluate polynomial p(x) = sum_{j=0}^{n} a_j*x^j using fast_power method
    
    input:
      * a_coeff: List with polynomial cofficients a_0,a_1, ..., a_n
      * x: value at which to evaluate the polynomial
    output:
      value of polynomial p(x)
    '''
    degree = len(a_coeff)-1     # The list a_coeff contains n+1 elements, where n is the degree 
    poly = a_coeff[0]    
    for j in range(1,degree+1):   # Loop from 1 to n
        aj_xj = a_coeff[j]
        xj = model_solutions.fast_power(x,j)
        poly += a_coeff[j]*xj # Update polynomial
    return poly
````

## Proof of polynomial complexity theorem (not examinable)
To prove that $f(n)=b_0+b_1 n+ b_2 n^2 + \dots + b_p n^p = \Theta(n^p)$ for $b_p>0$ we show that $f(n) = \mathcal{O}(n^p)$ and $f(n) = \Omega(n^p)$.

### Step 1: $f(n)=\mathcal{O}(n^p)$
For $n\ge 1$ we have that $n^q\le n^p$ for all $q=0,1,2,\dots,p$. Since for any real number $x\le |x|$ this gives

$$
\begin{aligned}
f(n) = b_0+b_1 n+ b_2 n^2 + \dots + b_p n^p &\le |b_0|+|b_1| n+ |b_2| n^2 + \dots + b_p n^p \\
&\le |b_0|n^p+|b_1| n^p+ |b_2| n^p + \dots + b_p n^p\\
&= \left(|b_0|+|b_1|+ |b_2| + \dots + b_p\right)n^p
\end{aligned}
$$

Since the expression in brackets is a positive constant, we find that $f(n)=\mathcal{O}(n^p)$ with $n_0=1$ and $c=|b_0|+|b_1|+ |b_2| + \dots + b_p$.
### Step 2: $f(n)=\Omega(n^p)$
First, factor out $b_pn^p$ to obtain

$$
\begin{aligned}
f(n) = b_0+b_1 n+ b_2 n^2 + \dots + b_p n^p &= b_p n^p\left(1+\frac{b_{p-1}/b_p}{n}+\frac{b_{p-2}/b_p}{n^2}+\dots+\frac{b_1/b_p}{n^{p-1}}+\frac{b_0/b_p}{n^{p}}\right)\\
&= b_pn^p \left(1-S(n)\right)\qquad\text{with}\quad S(n) := \sum_{k=1}^{p} \frac{-b_{p-k}/b_p}{n^k}.
\end{aligned}
$$

We now show that there is an $n_0$ such that $S(n)\le\frac{1}{2}$ for $n\ge n_0$. For this set

$$
n_0 = \max\left\{2\frac{|b_0|+|b_1|+|b_2|+\dots+|b_{p-2}|+|b_{p-1}|}{b_p},1\right\}
$$

For $n\ge n_0$ we have that $n\ge 1$. Since further $-x \le |x|$ for all real number $x$ this gives

$$
\frac{-b_{p-k}/b_p}{n^k} \le \frac{|b_{p-k}|/b_p}{n^k} \le \frac{|b_{p-k}|/b_p}{n}
$$

Applying this ineqality to each of the terms in the sum $S(n)$ and using the definition of $n_0$ we get

$$
S(n) = \sum_{k=1}^{p} \frac{-b_{p-k}/b_p}{n^k} \le \frac{|b_0|+|b_1|+|b_2|+\dots+|b_{p-2}|+|b_{p-1}|}{b_p n} \le \frac{1}{2}\frac{n_0}{n}\le \frac{1}{2}\qquad\text{for $n\ge n_0$}.
$$

This implies that for $n_0\ge n$ we have that

$$
f(n) = b_p n^p \left(1-S(n)\right) \ge \frac{1}{2}b_p n^p
$$

and hence $f(n)=\Omega(n^p)$ with $c=\frac{1}{2}$ and $n_0$ as defined above. Taking everything together we deduce that $f(n)=\Theta(n^p)$, which completes the proof.

### Proof that polynomial evaluation with $\text{fast_power}$ is $\Omega(n\log_2(n))$ (not examinable)
The number of elops required to compute $x^k$ with the fast_power method is at least $2(\log_2(k)-1)$. Taking into account the $n-1$ multiplications and additions that are required in the naive polynomial evaluation, we can bound the total number of elops required from below as

$$
E_{\text{naive_fast_power}}(n) \ge 2(n-1) + 2\sum_{k=2}^{n}\left(\log_2(k)-1\right) = 2\sum_{k=1}^{n} \log_2(k) 
\qquad(\ddagger)
$$

Next, let us look at $n!=1\cdot 2 \cdot 3 \cdots (n-1)\cdot n$ again. The number of factors which are at least as large as $n/2$ can be bounded from below by $n/4$. Hence

$$
n! \ge \left(\frac{n}{2}\right)^{n/4} \quad\Rightarrow \quad \sum_{k=1}^{n} \log_2(k) \ge \frac{n}{4}\log_2\left(\frac{n}{2}\right) = \frac{n}{4}\left(\log_2(n)-1\right)
$$

Using this and Eq. ($\ddagger$), we can derive a lower bound on $E_{\text{naive_fast_power}}(n)$:

$$
\begin{aligned}
E_{\text{naive_fast_power}}(n) &\ge 2\sum_{k=1}^{n} \log_2(k) \\
&\ge \frac{n}{2}\left(\log_2(n)-1\right)\\
&= \frac{n}{4}\log_2(n) + \frac{n}{4}\underbrace{\left(\log_2(n)-2\right)}_{\text{$\ge 0$ for $n\ge 4$}}\\
&\ge \frac{n}{4}\log_2(n)\qquad\text{for $n\ge 4$}.
\end{aligned}
$$

From this we deduce that $E_{\text{naive_fast_power}}(n)=\Omega(n\log_2(n))$ with $c=\frac{1}{4}$ and $n_0=4$. Note, incidentially, that we have also shown that $\log_2(n!) = \Theta(n\log_2(n))$.

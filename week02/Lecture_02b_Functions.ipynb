## Motivation: summing squares
Let's go back to the example from the last lecture, where we computed the sum of the squares of all numbers between 0 and 7:

$$
\text{sum_sq} = \sum_{j=0}^{7} j^2
$$

The Python code for this is

```Python
sum_sq = 0
for j in range(8):
    sum_sq += j**2
```

What if we want to compute this sum with a general upper bound $n$? Mathematically, this defines a function $\text{sum_sq_general}(n)$ which takes the integer number $n$ as an input and returns the value of the sum

$$
\text{sum_sq_general}(n) = \sum_{j=0}^{n} j^2
$$

We want to ''package up'' this calculation as a Python function, so that we can calculate the sum of the squares of all integer between $0$ and, say, $n=65$ by just typing
```Python
sum_sq_65 = sum_sq_general(65)
print(sum_sq_65)
```

## Python functions
To achieve this, we define a function as follows:

````python
def sum_sq_general(n):
    sum_sq = 0
    for j in range(n+1):
        sum_sq += j**2
    return sum_sq
````

We can use (or call) this function by typing its name followed by the actual value of the number $n$ for which we want to calculate the sum, for example:

````python
sum_sq_general(7)
````

(if you get an error when executing this cell you probably forgot to execute the cell with the definition of `sum_sq_general()` first).

In general, functions are defined with the `def` keyword, followed by the name of the function and the argument  in brackets (in computer science, the input is usually called ''argument'' or ''parameter'', but this is just terminology). As with for-loops, the definition finishes with a colon (`:`). Everything the function does is contained in the *function body* which needs to be indented. Anything that is not idented will not be part of the function. In our case the function body is
```Python
sum_sq = 0
for j in range(n+1):
    sum_sq += j**2
return sum_sq
```
The `return` keyword controls what is returned by the function. In our case we return the value of the variable `sum_sq`, which contains the sum. We can use whatever is returned by your function later in the code. For example, to compute the product of the two sums $\sum_{j=0}^{4} j^2$ and $\sum_{j=0}^{6} j^2$, we can write

````python
sum_1 = sum_sq_general(4)
sum_2 = sum_sq_general(6)
prod = sum_1*sum_2
print (prod)
````

Note that this code is much shorter and less error prone than writing down two separate for-loops to compute the two sums.

## More on Python functions
To talk about functions, some additional nomenclature is useful. In the definition of the function

```Python
def sum_sq_general(n):
```

`n` is called a *parameter* (also known as *formal argument* or *dummy argument*). In the function body this acts as a placeholder for whatever is passed to the function when it is called. In contrast, the concrete value `7` in the function call

```Python
sum_sq_general(7)
```

is called the *actual argument*. If you call the function this way, `n` will be replaced by 7 everywhere in the function body.

You can pass more than one parameter to a function. Imagine that instead of computing the squares of all numbers between $0$ and $n$, we want to compute the sum of the $\alpha$-th power of those numbers, i.e.

$$
\text{sum_pow_general}(n,\alpha) = \sum_{j=0}^{n} j^\alpha
$$

This is easily done in Python with the following function which gets passed the two parameters $n$ and $\alpha$:

````python
def sum_pow_general(n,alpha):
    sum_pow = 0
    for j in range(n+1):
        sum_pow += j**alpha
    return sum_pow
````

### Several return values
A function can also return more than one thing. Imagine that we want to add two fractions $\frac{a}{b}$ and $\frac{c}{d}$ and return the resulting fraction $\frac{f}{g}$ given by

$$
\frac{f}{g} = \frac{a}{b}+\frac{c}{d} = \frac{a\cdot d+b\cdot c}{b\cdot d}
$$

Here $a,b,c,d,f$ and $g$ are positive integers and we assume that $b$ and $d$ are not zero. To compute the numerator and denominator of $\frac{f}{g}$, we can write a Python function which is passed $a$, $b$, $c$ and $d$ and which returns $f$ and $g$ as follows:

````python
def add_fractions(a,b,c,d):
    f = a*d + b*c
    g = b*d
    return f, g
````

You can use this function to compute $\frac{3}{4} + \frac{7}{6} = \frac{46}{24}$:

````python
numerator, denominator = add_fractions(3,4,7,6)
print (numerator, denominator)
````

Functions can also not return anything. Consider for example the following function, which prints the sum of two numbers $a$ and $b$, but does not contain a return statement:

````python
def print_sum(a,b):
    print(a+b)
````

What happens if you try to assign a variable to the result of this function? Let's try it out:

````python
result = print_sum(2,3)
print(result)
````

Instead of throwing an error, Python happily executes the code, and returns the special value `None`. This is something to watch out for: if you forgot to add a `return` statement it will automatically return `None`, and this will often break any subsequent computations.

## Documenting functions
As with all our code, it is important that we document our functions. Inside the function, use `#` for comments as usual. Since functions are often used by other people (including your future self), for the user it is particularly important to know

* *what* a function does (not necessarily *how* it does it),
* what the input parameters are and
* what the function returns

This information, which describes the so-called *interface* of the function, can be summarised in a docstring, starting and ending with three quotation marks `'''`. Consider the following function for multiplying two fractions $\frac{a}{b}$ and $\frac{c}{d}$:

````python
def multiply_fractions(a,b,c,d):
    '''Multiply the two fractions a/b and c/d to obtain the fraction f/g
    
    input: a,b and c,d: integers representing a/b and c/d
    returns: integer f, g representing the fraction f/g
    '''
    f = a*c
    g = b*d
    return f, g
````

The nice thing about docstrings is that we can print them out with the `help()` function which we already used to read the documentation of built-in Python functions. For example:

````python
help(multiply_fractions)
````

When we give this function to somebody else this is really all they need to know to use it.

### Solution to questions

#### Simplified fraction
The following code returns the simplified fraction, cancelling out any common factors between numerator and denominator. Remember to load the `numpy` library before using the `np.gcd()` function.

````python
import numpy as np
def add_fractions_simplified(a,b,c,d):
    '''Add the two fractions a/b and c/d to obtain the fraction f/g; cancel common factors in f and g
    
    input: a,b and c,d: integers representing a/b and c/d
    returns: integer f, g representing the fraction f/g
    '''
    f = a*d + b*c
    g = b*d
    gcd_fg = np.gcd(f,g)
    return f//gcd_fg, g//gcd_fg
````

Note that I used `//` instead of `/` to divide the numbers. This will always work because by definition both $f$ and $g$ are multiples of $\text{gcd}(f,g)$.

#### The do_something() function
For any given number $a$ the code
```Python
def do_something(a):
    x = a//3
    y = a-3*(a//3)
    return y
```

computes the remainder of $a$ with respect to division by 3. In other words it finds the integer $r$ such that

$$
a = 3q + r\qquad \text{such that $r\in\{0,1,2\}$}.
$$

With this we have that $\text{do_something}(7)=1$ and $\text{do_something}(50)=2$.

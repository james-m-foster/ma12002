### &#9745; Task 1
To raise the matrix $A$ to some power we can repeatedly multiply the matrix $A$ with itself, using the `@` operator, starting with the identity matrix (we will look at a different, efficient method for computing matrix powers later in this course). A Python function for computing the number of all walks of length $L$ between two nodes $i$ and $j$ is given as follows:

````python
import numpy as np
def count_walks(i,j,L):
    '''Count the number of walks of length L between nodes i and j
    
    input
      i: index of starting node
      j: index of ending node
      L: length of walk
      
    returns:
      number of walks of length L between i and j
    '''
    A = np.array([[0,1,1],[1,0,2],[1,2,1]]) # adjacency matrix of the graph
    B = np.array([[1,0,0],[0,1,0],[0,0,1]]) # start with the identity matrix
    
    for k in range(L):
        B = A @ B

    return B[i,j]
````

Recall that a matrix can also be raised to some power with the `linalg.matrix_power()` method from the numpy library. If we use this, the `count_walks()` function can be simplified somewhat to (leaving out the docstring):

```Python
import numpy as np
def count_walks(i,j,L):
    A = np.array([[0,1,1],[1,0,2],[1,2,1]])
    B = np.linalg.matrix_power(A,L)
    return B[i,j]
```

We can now use our `count_walks()` function to compute the number of walks of length $2$ between vertices $0$ and $2$ as follows:

````python
print(count_walks(0,2,2))
````

This agrees with the manual calculation shown in the tickables.

### &#9745; Task 2

To count all walks that are no longer than $8$ we can use the following for-loop (recall that the bound in the for-loop is always one larger than the largest value we want to consider):

````python
total_walks = 0

for L in range(9):
    total_walks += count_walks(0,2,L)

print(total_walks)
````

Hence, we conclude that the number number of walks with length not larger than $8$ is $4411$.

## Further things to try out
For $L=0$ the code returns $1$ if $i=j$ and it returns $0$ if $i\ne j$. This makes sense if we define that there is one walk of length $0$ from each node to itself and there are no walks of length $0$ from a node to different node.

Here is some code which computes the sum $\sigma(n)$:

````python
n = 100
sigma = 0.0

for j in range(n+1):
    sigma += 4*(-1)**j/(2*j+1)

print(sigma)
````

If we set $n$ to a large value, e.g. $n=100000$ we find that

$$
\sigma(100000) = 3.1416026534897203
$$

which is close to 

$$
\pi\approx 3.141592653589793.
$$

We conclude that the sum converges to $\pi$. However, convergence is quite slow, as the example shows the fourth digit after the decimal point is still wrong if $n=100000$. There are more efficieny ways of approximating $\pi$ with a sum, see for example the discussion of the [Chudnovsky algorithm and Ramanujan's $\pi$ formulae](https://en.wikipedia.org/wiki/Chudnovsky_algorithm).

## Summing square numbers
In the last notebook we calculated the sum of all square numbers between 0 and 7,

$$
0^2 + 1^2 + 2^2 + 3^2 + 4^2 + 5^2 + 6^2 + 7^2 = \sum_{j=0}^{7} j^2
$$

Recall that in Python we can do this and save the result in a variable called `sum_sq` by writing

```Python
sum_sq = 0**2 + 1**2 + 2**2 + 3**2 + 4**2 + 5**2 + 6**2 + 7**2
```

This is quite nice, but what if we want to sum the squares of all numbers between 0 and 99? Or between 0 and 999? We'll have to type a lot to do this, and it is quite likely that we will make a mistake. It turns out, however, that it is very easy to compute this sum in Python without much typing.

## Pseudocode
To see this, let's first think about how we would compute the sum ''by hand''. We would start with $0^2$, then add $1^2$, followed by adding $2^2$, followed by adding $3^2$ and so on. We can express this procedure in the following *algorithm*:

### Algorithm: Sum squares of all numbers between 0 and 7
**Output**: Value of the sum $\sum_{j=0}^{7} j^2$
1. $\text{sum\_sq}\mapsto 0$
1. **for** $j=0,1,2,\dots,7$ **do**
1. $~~~~$ $\text{sum\_sq}\mapsto \text{sum\_sq}+j^2$
1. **end for**

Let's look at this more closely:

* in the first line we initialise the variable $\text{sum\_sq}$ to zero
* in the following lines we loop over all values of the variable $j$ between 0 and 7. For every value we increment the sum by $j^2$. This is expressed by the line $\text{sum\_sq}\mapsto \text{sum\_sq}+j^2$ which can be translated as *'' take the current value of the variable $\text{sum_sq}$, add $j^2$ to this and assign the variable $\text{sum\_sq}$ to this new value''*


The algorithm above is written in so-called *pseudocode*. This is not Python code yet, but it is a more formal way of expressing the solution to our mathematical problem. Once we've got the pseudocode right, translating it to Python is much easier.

In general, when solving a mathematical problem with programming, we will usually first try to express the solution procedure in pseudocode, which is the translated to Python code. Roughly speaking, the process looks like this:

$$
\text{Mathematical problem} \rightarrow \text{Solution idea} \rightarrow \text{Pseudocode} \rightarrow \text{Python code}
$$

It is very important not to skip any of those steps. For example, if we get the pseudocode wrong, we can not expect to write correct Python code.

We are often interested in how long it takes to execute a certain piece of code. As we will see later, pseudocode will allow us to do this.

## For-loops in Python
So now that we have our pseudocode, how do we translate the pseudocode to Python? Here is the solution:

````python
sum_sq = 0 # Initialise the sum to zero
for j in range(0,8):
    sum_sq = sum_sq + j**2 # increment the sum, read as "Add j^2 to SumSq and assign SumSq to this new value"
    
print(sum_sq)
````

The hash symbol `#` starts a comment, anything after the `#` will be ignored by Python (more on this below). From the code we deduce that the general syntax of a for-loop which iterates over the values $m,m+1,\dots,n-1$ is

```Python
for j in range(m,n):
```

(don't forget the colon '':'' at the end of the line). **Note that the second number passed to `range()` is one larger than the largest value that we want to loop over.** Here $j$ is called the *loop variable* and everything that is done in each iteration of the loop is called the *loop body*. In the code for summing up the squares of all numbers between 0 and 7, the loop body consists of just one line, namely

```Python
sum_sq = sum_sq + j**2
```

Python identifies the loop body by looking at the indentation level. You need to indent all lines by exactly the same number of spaces, usually four.

To increment a variable $a$ by some amount $b$, we can also write `a += b` which is exactly the same as `a = a + b`. Convince yourself that this is true by replacing `sum_sq = sum_sq + j**2` by `sum_sq += j**2` in the loop body in the code above.

Indentation is important in Python. Consider the following code prints out $j$ and $j^2$ for all values of $j$ between $0$ and $3$. Run the code and make sure you understand what it prints out.

````python
for j in range(0,4):
    print (j)
    print (j**2)
````

Now modify the code by un-indenting the final line

```Python
for j in range(0,4):
    print (j)
print (j**2)
```

and run it again. Do you understand the output? What is the loop body in the original code and in the modified code?

### Nesting loops
You can have a loop inside a loop, this is known as a *nested loop*. For example, to sum all values of $j\cdot k^3$ for $j=1,2,\dots,4$ and $k=2,\dots,5$ we can do this:

````python
another_sum = 0
for j in range(1,5):
    for k in range(2,6):
        another_sum += j*k**3
print (another_sum)
````

Observe that now the loop body of the inner loop over $k$ (`another_sum += j*k**3`) is indented by $8 = 2\times 4$ spaces. The loop body of the outer loop consists of the two lines

```Python
for k in range(2,6):
    another_sum += j*k**3
```

and it is indented by four spaces.

What happens if you indent the final line `print (another_sum)` by four or eight spaces? Modify the code and make sure you understand the output.

You can create loop nests with an arbitrary number of levels, for example:
```Python
for j in range(0,4):
    for k in range(4,6):
        for ell in range(8,12):
            # do something
    for m in range(2,6):
        # do something else
```

### General syntax of for-loops
In the examples above we always counted up in steps of $1$. We can also count in steps of a different size, say $k>0$. The following loop iterates over all values $j=m,m+k,m+2k,\dots$ such that $m+rk<n$, where $r$ is a non-negative integer; the number $k$ is also called the *increment*:

```Python
for j in range(m,n,k):
    # do something
```

To count down, simply pick a negative increment $k<0$. In this case $j$ will take take on all values $j=m,m+k,m+2k,\dots$ such that $m+rk>n$ where again $r$ is a non-negative integer.

Finally, to make our live easier, if we only pass *one* argument to `range()`, i.e. write

```Python
for j in range(n):
    # do something
```

this is exactly the same as writing

```Python
for j in range(0,n):
    # do something
```

As you can see from this example, Python inherently starts counting from zero.

## Comments and good style
Let's have another look at comments. As explained above, anything that starts with `#` is ignored by Python, for example if you write 

```Python
# This program increments a number
a += 1 # increment a by one
```

Python only sees

```Python
a += 1
```

So why should we bother writing comments? The reason is that the code will be looked at by other humans very frequently, and they need to understand it to use it in a sensible way. One of these other humans could be your friend or co-worker, but it might also be your future self who is trying to understand what you did a few weeks ago. As you will find out while working through this course, it can be surprisingly hard to make sense of code you wrote a few days ago. This can be avoided by commenting all code. When doing this, it helps to keep the following in mind:

**The purpose of comments is to add meaning to code**. Good comments should ...

* ... provide additional information
* ... explain the rationale (the 'why') behind some code we wrote
* ... be neither too short nor too verbose

Consider the following two examples:

##### good comment:
```Python
area = a*b # Calculate area of rectangle with sides a and b
```
The comment makes it clear that `a` and `b` represent the sides of a rectangle and that `area` is the resulting area.

##### bad comment:
```Python
area = a*b # Set area to a times b
```
...or even worse:
```Python
area = a*b # area = a*b
```
These comments does not add any information. They do not explain what the variables `a`, `b` and `area` represent.

### What is good style?
Comments help to make our code easier to understand, but there are a couple of other things we can do to make our code more readable, for example by formatting it properly. As a rough rule of thumb, about half of our code should consist of comments. There are a few other things we can do to make your code as readable as possible. Python already forces us to indent your code, for example when writing for-loops. In addition, we should use descriptive variable names such as `area`, `volume` or `sum_of_cubes`. It is best to restrict one-letter variables names such as `j`, `k` etc. to loop variables (but avoid lower case `l` which is very hard to distinguish from the number `1` - use `ell` instead). Excessively long variable names such as `sum_of_all_cubes_between_13_and_76` should also be avoided.

### The Python PEP 8 style guide
To ensure that Python code is written with good style, a set of guidelines has been established and published as a so-called *Python Enhancement Proposal (PEP)*. Those guidelines are known as the [PEP 8 Style Guide](https://www.python.org/dev/peps/pep-0008/), and you are strongly encouraged to follow them as far as possible. A small number of style points will be awarded for well documented and readable code in the coursework assignments.

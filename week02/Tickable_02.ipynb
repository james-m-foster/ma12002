## Graphs and walks
Before we can start we first need some mathematical definitions.

A **graph** is a set of $n$ *nodes*, numbered from $0$ to $n-1$, which are connected by *edges*.

Consider the following example:

!["Graph with three nodes"](./figures/graph.svg "Graph with three nodes")

This graph has $n=3$ nodes. Both nodes $1$ and $2$ are connected to node $0$ by exactly one edge. Nodes $1$ and $2$ are connected by two edges, and finally node $2$ is connected to itself by one edge (shown as a little loop).

A *walk* of length $L$ between nodes $i$ and $j$ is a way of traversing the graph such that
* we start at node $i$ and finish at node $j$
* we traverse exactly $L$ edges.

Here is an example of a walk of length $5$ between nodes $0$ and $2$:

!["Walk of length 5"](./figures/graph_walk_length5.svg "Walk of length 5")

In this case we traverse the nodes in the following order:

$$
0 \rightarrow 1 \rightarrow 2 \rightarrow 0 \rightarrow 2 \rightarrow 2
$$

Note that it is possible to walk along the same edge twice, and that edges can be traversed in any order.

The mathematical question that we will try to answer in this notebook is the following:

**Given two nodes $i$ and $j$, how many walks of length $L$ exist between these two nodes?**

If $L$ is small, this question can be answered quite easily by simply looking at the graph and writing down all possible walks. For example, there are three walks of length $2$ between $0$ and $2$, and they are shown in the following diagrams:

!["Walks of length 2 between nodes 0 and 2"](./figures/graph_walks_length2.svg "Walks of length 2 between nodes 0 and 2")

However, if $L$ is large, counting the number of walks will be become increasingly difficult (can you count the number of walks of length $3$ between nodes $0$ and $1$?). 

Fortunately, there is a very elegant way of counting walks, which uses a connection between graphs and matrices.

## The adjacency matrix
For each graph with $n$ nodes we can write down a so-called *adjacency matrix*. This is an $n\times n$ matrix with the following property: For all $i,j = 0,1,\dots n-1$ the matrix entry $A_{i,j}$ is the number of edges between nodes $i$ and $j$. For example, for the graph above the adjacency matrix would be

$$
A = \begin{pmatrix}
0 & 1 & 1 \\
1 & 0 & 2 \\
1 & 2 & 1
\end{pmatrix}.
$$

If we are interested in the number of walks of length $L$ between nodes $i$ and $j$, all we need to do is raise the adjacency matrix to the $L$-th power to obtain a new $n\times n$ matrix $B$, and look at the entry in row $i$ and column $j$ of the resulting matrix. In other words:

$$
\text{number of walks of length $L$ between nodes $i$ and $j$} = B_{ij}\qquad \text{where}\; B = A^L = \underbrace{A\circ A \circ \dots \circ A}_{\text{$L$ times}}.
$$

The (non-examinable) proof of this statement is given at the very end of this notebook.

For example, if $L=2$ then we find that

$$
B = A\circ A = \begin{pmatrix}
2 & 2 & \color{red}{3}\\
2 & 5 & 3\\
\color{red}{3} & 3 & 6
\end{pmatrix}.
$$

Looking at the entry $\color{red}{B_{02}=3}$, which is highlighted in red above, our formula predicts that there are three walks of length two between nodes $0$ and $2$. This agrees with what we found above.

## Proof (not examinable)
We use induction to prove the following statement:

*''The number of walks of length $L$ between nodes $i$ and $j$ is given by $\left(A^L\right)_{ij}$''*

To simplify notation, we will write $n_{\text{walks}}(i,j;L)$ for the number of walks of length $L$ between nodes $i$ and $j$ in the graph. Hence, for a given value of $L$ the statement that we want to prove is

$$
P(L): n_{\text{walk}}(i,j;L) = \left(A^L\right)_{ij} \quad\text{for all}\; i,j=0,1,\dots,n-1.
$$

#### Base case ($L=1$)
First, consider the case $L = 1$. In this case, the number of walks of length $1$ between $i$ and $j$ is simply given by the entry $A_{ij} = \left(A^1\right)_{ij}$ of the adjacency matrix, so the statement $P(1)$ is true, and in particular we have that $n_{\text{walk}}(i,j;1) = A_{ij}$.
#### Inductive step ($L\ge2$)
To walk from vertex $i$ to vertex $j$ in $L\ge 2$ steps, we first need to take one step (= a walk of length 1) from node $i$ to some node $k=0,1,\dots,n-1$ and then take a walk of length $L-1$ from node $k$ to node $j$.
This implies that

$$
n_{\text{walk}}(i,j;L) = \sum_{k=0}^{n-1} n_{\text{walk}}(i,k;1)\cdot n_{\text{walk}}(k,j;L-1)\qquad(\dagger)
$$

Now assume that $P(L-1)$ is true and hence $n_{\text{walk}}(\ell,m;L-1)=\left(A^{L-1}\right)_{\ell m}$ for all $\ell,m=0,1,\dots,n-1$. Since we also know that $n_{\text{walk}}(\ell,m;1) = A_{\ell m}$, Eq. ($\dagger$) is equivalent to

$$
n_{\text{walk}}(i,j;L) = \sum_{k=0}^{n-1} A_{ik}\left(A^{L-1}\right)_{kj} = \left(A^L\right)_{ij}
$$

where we have used the definition of matrix multiplication in the final step. Hence $P(L-1)\Rightarrow P(L)$, which completes the proof by induction.

## &#9745; Task 1
Recall the `n//2` is the integer part of $n/2$. Hence `2*(n//2)` is identical to `n` if and only if $n$ is even. A Python implementation of the pseudocode with error checking is given as follows:

```python
def dfactorial_iter(n):
    '''Iteratively compute double factorial n!! defined by
    
        n!! = 1*3*5*...*n for n even and n!! = 2*4*6*...*n for n odd with 0!! = 1.
    
    input:
      number n
    returns:
      double factorial n!
    '''
    if n < 0:
        raise Exception("n has to be non-negative")
    elif not (n == int(n) ):
        raise Exception("n has to be an integer")    
    elif n == 0:
        return 1
    elif n==2*(n//2):
        F = 1
        for j in range(2,n+1,2):
            F *= j
        return F
    else:
        F = 1
        for j in range(1,n+1,2):
            F *= j
        return F
```

We find for example that $5!!=5\cdot 3=15$, $6!!=6\cdot 4 \cdot 2 = 46$ and $13!!=13\cdot 11\cdot 9\cdot 7\cdot 5\cdot 3=135135$:

```python
print(dfactorial_iter(5))
print(dfactorial_iter(6))
print(dfactorial_iter(13))
```

As expected, the code aborts with an error message for invalid inputs:

```python
dfactorial_iter(-4)
```

## &#9745; Task 2
For the recursive implementation we need to cover the cases $n=0$ and $n=1$:

```python
def dfactorial_rec(n):
    '''Recursively compute double factorial n!! defined by
    
        n!! = 1*3*5*...*n for n even and n!! = 2*4*6*...*n for n odd with 0!! = 1.
    
    input:
      number n
    returns:
      nouble factorial n!
    '''
    if n < 0:
        raise Exception("n has to be non-negative")
    elif not (n == int(n) ):
        raise Exception("n has to be an integer")    
    elif (n == 0) or (n==1):
        return 1
    else:
        return n*dfactorial_rec(n-2)
```

As expected, the function returns the same result as `dfactorial_iter()`:

```python
print(dfactorial_rec(5))
print(dfactorial_rec(6))
print(dfactorial_rec(13))
```

### Tri-chess
Let $T(n)$ be the number of pairings in a Tri-chess tournament with $n$ players. When pairing up players, the first player has to pick two opponents. There are $n-1$ choices for the first opponent and $n-2$ choices for the second opponent. However, to avoid double counting we need to divide by two. After the first player has been paired up, the remaining $n-3$ players have to be organised into pairs of threes. There are $T(n-3)$ possibilities for this. Hence, the function $T(n)$ satisfies the following recursion formula:

$$
T(n) = \frac{1}{2}(n-1)(n-2)T(n-3)
$$

If $n=3$ there is only one triple pairing, so $T(3)=1$. Using this recursive definition, we arrive at the following recursive Python function for computing $T(n)$:

```python
def tri_chess(n):
    '''Compute the number of triple-pairings in a tri-chess tournament
    
    input
      n: number of players (has to be a multiple of 3)
    returns
      number of possible triple-pairings T(n)
    '''
    if n < 0:
        raise Exception("n has to be positive")
    elif not (n == 3*(n//3) ):
        raise Exception("n has to be a multiply of 3")    
    elif n == 3:
        return 1
    else:
        return (n-1)*(n-2)//2*tri_chess(n-3)
```

There are $15400$ triple-pairings in a tri-chess tournament with $n=12$ playes, as the following code demonstrates:

```python
tri_chess(12)
```

### Improved iterative double factorial
The iterative implementation can be simplified by counting down from $n$. If we do this, we no longer have to distinguish between even and odd values of $n$:

```python
def dfactorial_iter_improved(n):
    '''Iteratively compute double factorial n!! defined by
    
        n!! = 1*3*5*...*n for n even and n!! = 2*4*6*...*n for n odd with 0!! = 1.
    
    input:
      number n
    returns:
      nouble factorial n!
    '''
    if n < 0:
        raise Exception("n has to be non-negative")
    elif not (n == int(n) ):
        raise Exception("n has to be an integer")    
    elif n == 0:
        return 1
    else:
        F = 1
        for j in range(n,0,-2):
            F *= j
        return F
```

Obviously, this does not change the results:

```python
print(dfactorial_iter_improved(5))
print(dfactorial_iter_improved(6))
print(dfactorial_iter_improved(13))
```

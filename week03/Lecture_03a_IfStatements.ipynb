## Another look at the sum of square numbers
In the last notebook we wrote a function to compute the sum of the squared of all integers between zero and some upper bound $n$:

$$
\text{sum\_sq\_general}(n) = \sum_{j=0}^{n} j^2
$$

Imagine that we now ask a different question:

*''for some given $Q\ge 0$, what is the smallest integer number $n$ such that $\sum_{j=0}^{n} j^2 \ge Q$?''*

In mathematical notation we want to find the number $n_{\min}(Q)$ defined as

$$
n_{\min}(Q) = \arg\\min_{n\in\mathbb{N}} \left\{\sum_{j=0}^{n}j^2 \ge Q\right\}
$$

We can write down an algorithm for finding $n_{\min}$ in pseudocode

### Algorithm: Compute smallest $n$ such that $\sum_{j=0}^{n}j^2 \ge Q$
**Input**: Value of $Q\ge 0$
**Output**: $n_{\min}(Q)$
1. **for** $n=0,1,2,\dots,Q$ **do**
1. $~~~~$ **if** $\text{sum\_sq\_general}(n)\ge Q$ **then**
1. $~~~~~~~~$ **return** $n$
1. $~~~~$ **end if**
1. **end for**

Note that
* the upper loop bound can be set to $Q$ since $n_{\min}(Q)\le Q$ (you should convince yourself this is true)
* instead of re-implementing the for-loop we are using the function ``sum_sq_general`` that we already wrote. Re-using existing functions is good practice, helps to make the code modular, i.e. break it up into small parts with a well-defined purpose.

## Python implementation
The Python implementation of the algorithm looks like this (the ``sum_sq_general`` function from the previous notebook needs to be redefined so that we can use it here):

````python
def sum_sq_general(n):
    '''Compute sum of all squares between 0 and n
    
    input: upper bound n
    output: sum 0^2 + 1^2 + 2^2 + ... + n^2
    '''
    sum_sq = 0
    for j in range(n+1):
        sum_sq += j**2
    return sum_sq

def upper_bound(Q):
    '''Compute find the smallest n such that 0^2 + 1^2 + ... + n^2 >= Q 
    
    input: non-negative integer 
    output: minimal value of n
    '''
    for n in range(Q+1):
        if sum_sq_general(n) >= Q:
            return n
````

As with for-loops and function definitions, the body of the if-statement (i.e. the piece of code that should be executed if the condition is true) has to be indented, and there is a colon (`:`) at the end of the if-statement. Incidentally, the Python code also confirms that we can call a function defined by us (``sum_sq_general()``) from another function (``upper_bound()``) without problems.

### Extra challenge
Can you find a way of computing $n_{\min}$ without calling the function ``sum_sq_general()`` in every iteration? What are the advantages and disadvantages of the two different approaches?

The following code computes $n_{\min}$ directly, using just a single for loop:
```Python
def upper_bound_nocall(Q):
    '''Compute find the smallest n such that 0^2 + 1^2 + ... + n^2 >= Q 
    without calling the function sum_sq_general()
     
    input: non-negative integer 
    output: minimal value of n
    '''
    sum_sq = 0
    for n in range(Q+1):
        sum_sq += n**2
        if sum_sq >= Q:
            return n
```
As we will see later, this code is actually more efficient for large values of $Q$, since we do not have to repeatedly compute the partial sums $\sum_{j=0}^n j^2$. However, one might argue that the code that uses a call to ``sum_sq_general()`` is easier to read since the line

``if sum_sq_general(n) >= Q:``

is very descriptive and easy to interpret. Readability of code can be more important that elegance or speed. Is this case re-using the function `sum_sq_general` might look like an overkill, but imagine that you are using a function which computes something more complicated. Simply using an already existing function will reduce a lot of code duplication and it will make it much easier to find errors in the code.

## General syntax of if-statements in Python
In addition to telling Python what we want to do if a condition is satisfied, we can also specify what to do if it **not** true. Imagine, for example, that we want to print out `positive` if a given number is positive and `negative` if the number is negative. This can be done by writing

````python
some_number = 4
if some_number > 0:
    print('positive')
else:
    print('negative')
````

Observe that there is again a colon (`:`) after the keyword `else`, and that the body of the else-statement (the code we want to execute if `some_number>0` is not true) must be indented. Enclosing the words `positive` and `negative` with quotation marks `'` defines a so-called string, more about this later (try leaving out the quotation marks and see what happens. Do you see what Python is trying to do here?).

If-statements can be nested in the same way as for-loops, but you must make sure that the indentation is correct. For example, imagine that we want to print `positive` if the number is larger than zero and `very positive` if it is larger than 1000. This could done as follows:

````python
some_number = 4000
if some_number > 0:
    if some_number > 1000:
        print('very positive')
    else:
        print('positive')
else:
    print('negative')
````

It turns out there is a more elegant way of doing this by using the `elif` statement:

````python
some_number = 400
if some_number > 1000:
    print('very positive')
elif some_number > 0:
    print('positive')
else:
    print('negative')
````

So Python first checks whether the first condition (`some_number > 1000`) is true. If this is not the case, it checks the condition in the `elif` statement, i.e. it checks whether the number is at least larger than 0. If none of those two conditions are satisfied it will execute the code in the body of the `else` statement. You can have more than one `elif` clause. The most general syntax of an if-statement is this

```Python
if condition_1
    # do something if condition_1 is true
elif condition_2:
    # do something if condition_2 is true
    # (and condition_1 is false)
elif condition_3:
    # do something if condition_3 is true
    # (and both condition_1 and condition_2 are false)
    # ...
elif condition_n:
    # do something if condition_n is true
    # (and condition_1, condition_2, ..., condition_{n-1} are all false)
else:
    # do something if condition_1, condition_2, ..., condition_n are all false

```

Conditions can be more complicated. Imagine that we want to print `apple` if a number is contained in the interval $[3,8)$ and `pear` if it is not. For this observe that

$$
x \in [3,8)\qquad\Leftrightarrow\qquad(x\ge 3)\;\text{and}\;(x<8)
$$

and write

````python
some_number = 4
if (some_number >= 3) and (some_number <8):
    # print 'apple' if some_number is contained in the intervale [3,8)
    print ('apple')
else:
    print ('pear')
````

Each of the statements in round brackets can be either true of false, and we can use the keywords `and`, `or` and `not` to combine them. Since mathematically

$$
x \in [3,8)\qquad\Leftrightarrow\qquad(x\ge 3)\;\text{and}\;(x<8)\qquad\Leftrightarrow\qquad
\text{not}\;\left((x < 3) \;\text{or}\; (x\ge 8)\right)
$$

an equivalent implementation is this:

````python
some_number = 4
if not ((some_number < 3) or (some_number >= 8)):
    # print 'apple' if some_number does not lie outside the intervale [3,8)
    print ('apple')
else:
    print ('pear')
````

As in the mathematical expressions, we need to make sure that we use brackets in the correct place:


Logically the statement 

$$
\text{not}\;\left((x < 3) \;\text{or}\; (x\ge 8)\right)
$$

is not the same as
$$
\left(\text{not}\;(x < 3)\right) \;\text{or}\; (x\ge 8)
$$

(Can you see why the latter statement is equivalent to $x\ge 3$?).

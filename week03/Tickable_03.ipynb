## The double factorial
for all non-negative integers $n\in\mathbb{N}_0$ the double factorial $n!!$  is defined as

$$
n!! = \text{dfactorial}(n) = \begin{cases}
1\cdot 3\cdot 5 \cdot \cdot \cdot (n-2)\cdot n & \text{for $n$ odd}\\
2\cdot 4\cdot 6 \cdot \cdot \cdot (n-2)\cdot n & \text{for $n$ even}\\
1 & \text{for $n=0$}
\end{cases}
$$

For example, we have that $7!!=1\cdot 3\cdot 5\cdot 7=105$ and $8!!=2\cdot 4\cdot 6\cdot 8=384$. In this tickale we will look at two different implementations of the double factorial.

## Iterative implementation
$n!!$ can be computed *iteratively* by using for-loops as shown in the following algorithm:
### Algorithm (iterative method)
**Input** Non-negative number $n$, **Output** double factorial $n!!$

1. $~~~~$ **return** $1$

1. $~~~~$ Set $F\mapsto 1$
1. $~~~~$ **for** $k=2,4,6,\dots,n-2,n$ **do**

1. $~~~~$ **return** $F$
1. **else**
1. $~~~~$ Set $F\mapsto 1$
1. $~~~~$ **for** $k=1,3,5,\dots,n-2,n$ **do**
1. $~~~~~~~~$ $F\mapsto F\cdot k$
1. $~~~~$ **end for**
1. $~~~~$ **return** $F$
1. **end if**

## Recursive implementation
It turns out that there is another ways of implementing the double factorial. For this, observe that the function $\text{dfactorial}(n)$ satisfies the following recursion relation:

$$
\text{dfactorial}(n) = \begin{cases}
n\cdot \text{dfactorial}(n-2) & \text{for $n\ge 2$}\\
1 & \text{for $n=0$ and $n=1$}
\end{cases}
\qquad(\dagger)
$$

As in the lectures, we can use this fact to write a *recursive* Python function for evaluating $n!!$.

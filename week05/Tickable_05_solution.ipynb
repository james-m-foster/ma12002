## &#9745; Task 1
The function `fib_rec_mat_vec()` can be implemented as follows:

```python
import numpy as np
def fib_rec_mat_vec(n):
    '''Compute the n-th Fibonacci vector (F_n, F_{n-1})

    input: Number n
    output: n-th Fibonacci vector (F_n, F_{n-1})
    '''
    if not (n == int(n)):
        raise Exception('Fibonacci number F_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Fibonacci number F_n not defined for negative values')
    elif (n == 0):
        return np.array([0,1])
    else:
        A_fib = np.array([[1,1],[1,0]])
        return A_fib@fib_rec_mat_vec(n-1)
```

## &#9745; Task 2
Recall the `fib_rec_mat_vec()` computes the vector $\boldsymbol{f}_n=\begin{pmatrix}F_n\\F_{n-1}\end{pmatrix}$ Based on this, we can implement a function `fib_rec_mat()` which extracts extracts the first entry of the vector $\boldsymbol{f}_n$ and hence returns $F_n$:

```python
def fib_rec_mat(n):
    '''Compute the n-th Fibonacci number by extracting the first component
    of the n-th Fibonacci vector (F_n, F_{n-1})

    input: Number n
    output: n-th Fibonacci number F_n
    '''
    if not (n == int(n)):
        raise Exception('Fibonacci number F_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Fibonacci number F_n not defined for negative values')
    f_fib = fib_rec_mat_vec(n)
    return f_fib[0]
```

The functions above check whether the input is an integer number (`n==int(n)`) and whether it is non-negative. If one of these conditions is violated they raise an exception, which triggers a Python error message.

Although both codes use recursion, the main difference is that `fib_rec_mat()` calls itself only *once* whereas `fib_rec()` calls itself *twice*. This explains the exponential explosion in runtime for `fib_rec()`; the runtime for `fib_rec_mat()` only grows linearly with $n$. We will come back to this later, but for the moment observe that recursion leads to a rapid growth in runtime only under certain conditions. In other words, recursion is not always bad.

### Challenge
The matrix we are looking for is

$$
A_{\text{trib}} = \begin{pmatrix}
1 & 1 & 1 \\
1 & 0 & 0 \\
0 & 1 & 0
\end{pmatrix}
$$

To see this, write down the matrix-vector product $A_{\text{trib}}\boldsymbol{t}_{n-1}$ componentwise:

$$
\begin{pmatrix}
1 & 1 & 1 \\
1 & 0 & 0 \\
0 & 1 & 0
\end{pmatrix}
\begin{pmatrix}
T_{n-1} \\ T_{n-2} \\ T_{n-3}
\end{pmatrix}
=
\begin{pmatrix}
T_{n-1} + T_{n-2} + T_{n-3} \\
T_{n-1} \\
T_{n-2}
\end{pmatrix}
=
\begin{pmatrix}
T_n \\ T_{n-1} \\ T_{n-1}
\end{pmatrix}
$$

The final equality follows from the definition $T_n = T_{n-1} + T_{n-2} + T_{n-3}$ of the Tribonacci numbers

The recursion is implemented in the following two Python routines. First, we recursively compute $\boldsymbol{t}_n$ in essentially the same way as $\boldsymbol{f}_n$ in `trib_rec_mat_vec()`. Next, we extract the first component of the vector $\boldsymbol{t}_n$ to obtain the $n$-th Tribonacci number with `trib_rec_mat()`.

```python
import numpy as np
def trib_rec_mat_vec(n):
    '''Compute the n-th Tribonacci vector (T_n, T_{n-1}, T_{n-2})

    input: Number n
    output: n-th Tribonacci vector (T_n, T_{n-1}, T_{n-2})
    '''
    if not (n == int(n)):
        raise Exception('Tribonacci number T_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Tribonacci number T_n not defined for negative values')
    elif (n == 0):
        return np.array([0,1,-1])
    else:
        A_trib = np.array([[1,1,1],[1,0,0],[0,1,0]])
        return A_trib@trib_rec_mat_vec(n-1)

def trib_rec_mat(n):
    '''Compute the n-th Tribonacci number by extracting the first component
    of the n-th Tribonacci vector (T_n, T_{n-1}, T_{n-2})

    input: Number n
    output: n-th Tribonacci number T_n
    '''
    if not (n == int(n)):
        raise Exception('Tribonacci number T_n not defined for non-integer values')
    elif (n < 0):
        raise Exception('Tribonacci number T_n not defined for negative values')
    t_trib = trib_rec_mat_vec(n)
    return t_trib[0]
```

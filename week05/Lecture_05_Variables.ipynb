## Assigning values in Python
Consider the following simple Python code:

````python
a = 1
b = a
print ("a = ",a," b = ",b)
a = 2
print ("a = ",a," b = ",b)
````

This code does what we would expect: after the first two lines, both `a` and `b` have the same value, namely 1, and this is confirmed by printing out those variables. As the final print statement demonstrates, setting `a` to 1 in the penultimate line has no effect on the variable `b`.

In this example both `a` and `b` are integers (type `int` in Python). Now let us do something very similar but with numpy arrays. More specifically:

1. create a numpy array `a_vec` which contains only zeros by writing `a=np.array([0,0,0])`
2. assign the new name `b_vec` to this array by writing `b_vec=a_vec`
3. change the first element of `a_vec` to 1.

The following Python code does this, while printing out the values of `a_vec` and `b_vec`:

````python
import numpy as np
a_vec = np.array([0,0,0])
b_vec = a_vec
print(" a_vec = ",a_vec," b_vec = ",b_vec)
a_vec[0] = 1
print(" a_vec = ",a_vec," b_vec = ",b_vec)
````

What's going on here? Modifying the variable `a_vec` seems to also change the value of `b_vec`! This seemingly odd behaviour can lead to unexpected consequences in our code if we are not careful. As we will see in the following, Python behaves in the seemingly odd way to save memory, i.e. to be more efficient.

### Copy by reference vs copy by value
The reason this happens here is because in contrast to the first example (where `a` and `b` were integers) the line `b_vec=a_vec` does not copy the *content* of `a_vec` to `b_vec`, but rather says *''use `b_vec` as a different name for `a_vec`''*. In computer science this is often called *copy (or assignment) by reference*. There are good reasons why Python behaves in this way: simple data types, such as integers, logicals and floats, only take up a small amount of memory, so copying them by value as in the statement `b=a` is cheap. In contrast, more complex data types such as numpy arrays can take up much more memory (imagine creating a $1000\times 1000$ matrix), so it is more efficient to just use the new variable `b_vec` as a reference pointing to the already existing `a_vec` in memory.

The behaviour of the `=` operator depends on the datatype, and the following table summarises this behaviour for the datatypes that we have met so far:

| datatype | copy by... |
| --- | --- |
| `int` | ...value |
| `float` | ...value |
| `bool` | ...value |
| `str` | ...value |
| `list` | ...reference |
| `numpy.ndarray` | ...reference |

In general, Python distinguishes between *immutable* data types such as `int`, `float`, `string` and `bool` and *mutable* data types such as `list` and `numpy.ndarray`.

### Enforcing assignment by value
While often copy by reference is ok as long as we are aware of the consequences, sometimes we might want to copy numpy arrays and lists by value. This can be done by using the name of the datatype as a function in the assignment. In the following example the lists `a_list` and `b_list` will be different (in the sense that they do not refer to the same memory):

```Python
a_list = [1,2,3]
b_list = list(a_list) # (instead of b_list = a_list)
```

This also works for two numpy arrays:

```Python
a_vec = np.array([1,2,3])
b_vec = np.array(a_vec)
```

The following code confirms that in this case copy is indeed by value:

````python
import numpy as np

# numpy arrays
a_vec = np.array([0,0,0])
b_vec = np.array(a_vec)   # Note that before we used b_vec = a_vec
print(" a_vec = ",a_vec," b_vec = ",b_vec)
a_vec[0] = 1
print(" a_vec = ",a_vec," b_vec = ",b_vec)
````

Changing the first element of `a_vec` in the penultimate lines does **not** modify the content of `b_vec`.

Finally, observe that while there is a difference between `b_list=a_list` and `b_list=list(a_list)` if `a_list` and `b_list` are lists, the following two statements are equivalent if `a_int` and `b_int` are of type `int`:

```Python
b_int = a_int
b_int = int(a_int)
```

The same applies for floats, strings and logicals, and the reason for this is that those are all immutable data types which are copied by value.

## Scope of variables
Having discussed assignment of variables, let us now look at how variables are used in functions. Consider the following Python code, which sets the variable `x` to 1 inside the function `set_x_to_one()` and then tries to assign `y` to `x+1` outside the function:

````python
a = 3

def set_x_to_a():
    '''Set x to a'''
    x = a

set_x_to_a()
    
y = x + 1
````

If we run this code, it will crash with an error: Python complains that it does not know what `x` is in the final line `y = x + 1`. This might be odd, since we set `x` to some value inside the function, which was called before the problematic statement `y = x + 1`! The problem is that the variable `x` that we defined *inside* the function is also only visible inside that that function. In computer science the region of visibility of a variable is called its *scope*, and in our example the variable `x` has so-called *local scope* where by ''local'' we mean ''local to the function'' which defines it.

In contrast, the variable `a` has *global* scope: it is visible everywhere, and we can use it inside the function without problems. Let us look at another example:

````python
a = 3

def set_a_to_x(x):
    '''Set a to x'''
    a = x

set_a_to_x(1)    
print (a)
````

Now this is also odd: Although we set the variable `a` to 1 inside the function, it appears that it actually has *not* changed at all! The reason for this is that as far as Python is concerned, there are actually *two* variables called `a`: One with *global* scope, defined in the very first line (`a = 3`) and another one with *local scope* defined inside the function (`a = x`). This becomes more transparent if we explicitly rename the variables as follows:

```Python
a_global = 3

def set_a_to_x(x):
    '''Set a to x'''
    a_local = x

set_a_to_x(1)    
print (a_global)
```

In some cases we *can* modify global variables inside a function, namely if they are of mutable type. Consider the following example, which increments the two elements of a vector by one and returns their product:

````python
import numpy as np

a_global = np.array([1,2])

def increment_mutiply():
    '''Increment and multiply'''
    a_global[0] += 1; a_global[1] += 1
    return a_global[0]*a_global[1]

print (increment_mutiply())
print (a_global)
````

Now the line `a_global[0] += 1; a_global[1] += 1` has modified the global variable `a_global`. This is usually a bad idea. Imagine, that we are just *using* the function in your code, for example because it was given to us by a friend. If we do not look at the details of the function it might become very hard to understand where our code is going wrong. We can try to fix this problem by copying the global variable `a_global` to a local variable `a_local` first:

````python
a_global = np.array([1,2])

def increment_mutiply():
    '''increment and multiply'''
    a_local = a_global
    a_local[0] += 1; a_local[1] += 1
    return a_local[0]*a_local[1]

print (increment_mutiply())
print (a_global)
````

Alas, this does not work either! The reason for this is that the line `a_local = a_global` copies by reference since both `a_global` and `a_local` are of mutable type. The problem can be easily fixed by replacing the problemenatic line by `a_local = np.array(a_global)` as in the following code:

````python
a_global = np.array([1,2])

def increment_mutiply():
    '''increment and multiply'''
    a_local = np.array(a_global)
    a_local[0] += 1; a_local[1] += 1
    return a_local[0]*a_local[1]

print (increment_mutiply())
print (a_global)
````

There is, however, still an issue with this function: what if we forget to define `a_global` before calling the function? In other words, if we give this function to someone else, how will the user know that they have to define `a_global` first, before calling the function? This issue can be avoided by simply passing the variable as an argument:

````python
def increment_mutiply(a_arg):
    '''increment and multiply
    
    input: numpy vector with two elements
    '''
    a_local = np.array(a_arg)
    a_local[0] += 1; a_local[1] += 1
    return a_local[0]*a_local[1]

a_global = np.array([1,2])
print (increment_mutiply(a_global))
print (a_global)
````

This is much better, since now the definition of the function (and the documentation!) tells us what the function expects. Note that the local variable is copied by value: `a_local = np.array(a_arg)`. If we write `a_local=a_arg`, the variable `a_local` will still be a reference to the dummy argument `a_arg`, which is in turn a reference to the actual argument `a_global`. Hence modifying `a_local` will still modify `a_global`! The following code demonstrates this:

````python
def increment_mutiply(a_arg):
    '''increment and multiply
    
    input and output: numpy vector with two elements
    '''
    a_local = a_arg # Copy by reference !
    a_local[0] += 1; a_local[1] += 1
    return a_local[0]*a_local[1]

a_global = np.array([1,2])
print (increment_mutiply(a_global))
print (a_global)
````

This is not necessarily wrong. There are cases where we might want to modify one of variables passed to a function, but if this is the case we should clearly alert the user to this by documenting this behaviour in the function's docstring as in the example above. For example, consider the following function which swaps the entries of a a list:

```Python
a_list = [1,2]
def swap(a):
    '''Swap the elements of a list with two entries
    
    input/output: list to be swapped'''
    tmp = a[0]
    a[0] = a[1]
    a[1] = tmp
swap(a_list)
```

### Encapsulation
I hope that all those issues have convinced you that using global variables inside functions is dangerous and can lead to very annoying bugs. The above examples might appear to be artificial, but it turns out that the issues they highlight are actually quite common. Those problems can be avoided by using a general design principle known as **encapsulation**.  In essence this means that a function should only communicate with the outside world through the arguments that are passed to it and through its return value.

In practice, this boils down to the following guidelines:

* **never ever** use global variables inside functions
* the computations inside a function must only depend on its arguments
* to return a value return it explicitly with the `return` statement
* if a dummy argument is of mutable type
   - be careful to copy it by value if it is modified inside the function and you do not want to change the actual argument
   - if you do want to modify a dummy argument, clearly document this in the function's docstring

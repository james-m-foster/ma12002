## Repeating things with while loops
Imagine that we want to find the largest integer number $n$ such that $n=2^k$ and $n<1000$ where $k$ is also an integer. In other words, we are looking for the largest power of 2 which is still smaller than 1000. To do this, we can repeatedly multiply the number 2 by itself as long as the resulting number is smaller than 1000. Here is some pseudocode for doing this:

### Algorithm: Find largest power of two smaller than $1000$
1. Set $n=1$
1. **while** $2n<1000$ **do**
1. $~~~~$ $n\mapsto 2\cdot n$
1. **end while**

The corresponding Python code looks like this:

````python
n=1
while 2*n<1000:
    n=2*n
print(n)
````

In Python the general syntax of a while loop is this:

```Python
while condition:
    do something
```

The repeats the instructions in the loop body (`do something`) as long as the `condition` is true. As in for loops, there is a colon at the end of the line that starts with `while`, and it is important to indent the loop body correctly.

Let's look at another example. Consider the following sequence:

$$
a_{k+1} = \begin{cases}
  \frac{1}{2}a_k & \text{if $a_k$ is even,}\\
  3a_k + 1 & \text{if $a_k$ is odd}
\end{cases}
$$

For a given $a_0$, we want to print all $a_0,a_1,\dots,a_n$ until $a_n=1$. We can not do this with a for loop, since we don't know $n$, but it is quite easy to do with a while-loop. To check whether a number is even we can look at the truth value of the statement `2*(a//2)==a`, which is only true if `a` is even.

Here is the code for $a_0=9$ (try running it with different starting values):

````python
a = 9
print (a)
while not (a == 1):
    if (2*(a//2)==a):
        a = a//2
    else:
        a = 3*a+1
    print (a)
````

### Infinite loops
In the example above it is actually not obvious that we will ever reach $a_n=1$, so we might end up printing out numbers $a_k$ forever. One of the dangers of while loops is that they might never finish. This can happen if the condition in the while-statement is always true. In this case we are stuck in an infinite loop. The simplest example is this:

````python
while True:
    x = 1
````

If we get stuck in an infinite loop, we will find that the execution of the cell doesn't stop. This is indicated by a star in the square brackets left of the cell, which looks like this: `In [*]`. If this happens, we can abort the calculation by selecting ''*Kernel*'' $\rightarrow$ ''*Interrupt*'' from the menu bar at the top of the notebook.

## When (not) to use a while loop
Keen observers might claim that we no longer need for loops since we can replace every for loop with a while loop. For example, let us imagine that we want to loop over the range 0 to 7, which is readily done with a for loop:

```Python
for k in range(8):
    # do something
```

This could also be done with a while loop as follows:

```Python
k = 0
while k < 8:
    # do something
    k = k + 1
```

Although this works, it is usually not a good idea. Observe that if we use a while loop, we need to explicitly take care of three things:

* we need to initialise k
* we need to write the actual while loop
* we have to remember to increment k in every iteration

First of all, compared to the for-loop above, we need to write more code. Secondly, it is very easy to make a mistake. What if we forget to initialise k or forget to increment it? Do we increment k before or after we do something in the loop? In fact, it is also very easy to accidentially write an infinite loop. Hopefully this convinces you that in the example above the for-loop is not only simpler, but also much less error-prone.

As a rule of thumb, you should always use a for loop if you can. Resort to using a while loop only of this is not possible. We have seen two examples in this lecture and will see more when implementing sorting algorithms.

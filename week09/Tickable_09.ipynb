## Background
Recall the definition of the longest increasing subsequence (see [Tickable 8](../week08/Tickable_08.ipynb) for more examples):
### Increasing subsequences
Let $a=a_0,a_1,\dots,a_{n-1}$ be a sequence of $n$ distinct integer numbers. A subsequence $a_j,a_{j+1},\dots,a_{k-1},a_k$ of $a$ with $0\le j \le k < n$ is called an **increasing subsequence** if $a_{\ell}<a_{\ell+1}$ for all $\ell=j,j+1,\dots,k-1$. The length of the increasing subsequence is $k-j+1$. We define that for each $0\le j<n$ the number $a_j$ forms an increasing subsequence of length $1$. Hence, every sequence $a$ trivially contains $n$ increasing subsequences of length $1$.

Our goal is to find the length $L_{\max}$ of the *longest* increasing subsequence of a given sequence $a$. 

## Divide and Conquer solution
The brute force algorithm described in [Tickable 8](../week08/Tickable_08.ipynb) has computational complexity $\Theta(n^3)$ (in the sense that it requires $\Theta(n^3)$ comparisons of subsequent numbers in the sequence), so it will be very slow for long sequences with $n\gg1$. To find a better solution, we first define

$$
n^* := \begin{cases}
\frac{n}{2} & \text{if $n$ is even}\\
\frac{n-1}{2} & \text{if $n$ is odd}.
\end{cases}
$$

Considering the element $a_{n^*}$, there are now two possibilities: either $a_{n^*}$ is part of the longest increasing subsequence of $a$ or it is not. In the latter case, the longest increasing subsequence of $a$ must be a subsequence of either $a_{\text{left}} := a_0,a_1,\dots,a_{n^*-1}$ or of $a_{\text{right}} := a_{n^*+1},a_{n^*+2},\dots,a_{n-1}$. In the former case we can find the longest increasing subsequence containing $a_{n^*}$ by first setting $j=n^*$ and decreasing $j$ as long as $a_j,\dots,a_{n^*}$ is an increasing subsequence. We then set $k=n^*$ and increase $k$ as long as $a_{n^*},\dots,a_k$ is an increasing subsequence. The length $L_{\text{centre}}$ of the longest increasing subsequence containing $a_{n^*}$ is then simply given by $k-j+1$. Finally, the length of the longest increasing subsequence of $a$ is obtained by taking the maximum of three numbers:
1. $L_{\text{centre}}$ as computed with the procedure just described
2. the length $L_{\text{left}}$ of the longest one-step subsequence of $a_{\text{left}}$
3. the length $L_{\text{right}}$ of the longest one-step subsequence of $a_{\text{right}}$

To compute the latter two numbers, we can apply our algorithm recursively to the two shorter lists $a_{\text{left}}$ and $a_{\text{right}}$. For large $n$ we have that $n^*\approx n/2$, so the method is applied to two lists of approximately half the length in the recursive call; this is why this is a Divide and Conquer approach.

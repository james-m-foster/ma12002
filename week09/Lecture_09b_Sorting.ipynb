## Sorting: some general considerations
Assume that we have a list of $n$ numbers $[a_0,a_1,a_2,\dots,a_{n-1}]$. For simplicity we will further assume that the numbers are mutually distinct, i.e. $a_j\ne a_k$ for $j\ne k$. We now want to sort those numbers in ascending order, for example:

$$
[7, 2, 5, 1, 10, 14, 19, 8, 3, 6] \mapsto [1, 2, 3, 5, 6, 7, 8, 10, 14, 19]
$$

In computer science, a lot of thought has been put into devising clever algorithms for doing this efficiently for very long lists of numbers. 

Of course, if we know how to sort numbers, we can also sort any other kind of objects for which we can as ''is $a$ smaller than $b$''? For example, we might want to sort a group of people by their first names

$$
[\text{Carlos}, \text{Susie}, \text{Adam}, \text{Beth}] \mapsto [\text{Adam}, \text{Beth}, \text{Carlos}, \text{Susie}]
$$

or by height as shown in the following figure:

![Sorting](./figures/SortedPeople.png "Sorting a group of people")

There are a lot of other applications that require sorting. Think for example of the music library on your smartphone or your address book. The latter might contain a relatively modest number of entries, but the address book of a large company can contain 10,000s of entries, so fast sorting algorithms are crucial.

## Insertion sort
Our list of numbers $[a_0,a_1,a_2,\dots,a_{n-1}]$ can be sorted in ascending order as follows:

1. Compare the second number to the first number, and swap them if the second number is smaller than the first number. After this, the first two numbers are sorted.
2. Now look at the third number, and compare it to the first two numbers (which are sorted after Step 1). If the third number is larger than the first two, there is no need to do anything. Otherwise, swap the third number with the second, and then possibly again with the first number until the three numbers are arranged in ascending order. After this, the first three numbers are sorted
3. Next, compare the fourth number to the first three (which are sorted after Step 2), and insert it at the correct position. After this, the first four numbers are sorted.
4. Repeat this with the remaining numbers until the end of the list is reached.

Let us formalise this idea as follows:

* A single number $a_0$ is trivially sorted
* Assuming the first $k$ numbers $a_0,a_1,a_2,\dots,a_{k-1}$ are sorted, we can arrange the first $k+1$ numbers $a_0,a_1,a_2,\dots,a_{k-1},a_{k}$ numbers in ascending order as follows: Take the number $a_{k}$ and compare it to the numbers $a_j$ with $j=k-1,k-2,\dots,1,0$ (note that we are counting down). If $a_j>a_{k}$, we need to insert $a_k$ somewhere before $a_j$, otherwise $a_{k}$ is at the correct position and we can stop since we inserted $a_{k}$ in the right place.

The following figure illustrates the process for $k=6$:

![Sorting with insertion sort](./figures/InsertionSort.svg "Sorting with insertion sort")

You can find a animation of this process on [this webpage](https://visualgo.net/en/sorting) (select ''INS'' for InsertionSort at the top, and then click on ''Go'' in the bottom left.

The method can be expressed as the following algorithm:

### Algorithm: Insertion Sort
**Input**: List of $n$ mutually distinct numbers $a=[a_0,a_1,a_2,\dots,a_{n-1}]$ **Output**: Rearrangement $b=[b_0,b_1,b_2,\dots,b_{n-1}]$ of the list in ascending order
1. Set $b=a$, i.e. set $b_k=a_k$ for $k=0,1,\dots,n-1$ (to avoid overwriting the original list)
1. **for** $k=1,2,\dots,n-1$ **do**
1. $~~~~$ Set $x\mapsto b_k$
1. $~~~~$ Set $j\mapsto k-1$
1. $~~~~$ **while** $j\ge 0$ and $b_j>x$ **do**
1. $~~~~~~~~$ $b_{j+1}\mapsto b_j$
1. $~~~~~~~~$ $j\mapsto j-1$
1. $~~~~$ **end while**
1. $~~~~$ $b_{j+1}\mapsto x$ (insert element at the correct place)
1. **end for**

The algorithm uses two loops. The outer loop over picks the $k$-th element, which is called $x$. In the inner loop over $j$, the number $x$ is then compared to the $k-1$ already sorted numbers $b_0,b_1,\dots,b_{k-1}$ and inserted in the correct place.

## Implementation in Python
Here is an implementation of the algorithm in Python:

```python
def insertion_sort(a):
    '''Sort a list using Insertion Sort
    
    input: list a
    output: rearrangement of the list a in ascending order
    '''
    b = list(a) # Create a copy to avoid overwriting the contents of a
    n = len(a)
    for k in range(1,n):
        x = b[k]
        j = k-1
        while (j >= 0) and (b[j]>x):
            b[j+1] = b[j]
            j = j-1
        b[j+1] = x
    return b
```

Observe that the inner loop is implemented as a while loop. This is appropriate here since we do not initially know over how many different values of $j$ we have to iterate.

Let's try out the Python code for the list $a = [7, 2, 5, 1, 10, 14, 19, 8, 3, 6]$:

```python
a = [7, 2, 5, 1, 10, 14, 19, 8, 3, 6]
insertion_sort(a)
```

## Complexity analysis
How long does it take to execute the Insertion Sort algorithm? To quantify this time, assume that the only operations that contribute to the runtime are the assignments $x\mapsto b_k$, $b_{j+1}\mapsto x$ and $b_{j+1}\mapsto b_j$, i.e. the first and last statements inside the for loop and the first statement inside the while loop. Let $E(n)$ be the number of times that we execute those three lines; the runtime will then be proportional to $E(n)$. Each of the statements $x\mapsto b_k$ and $b_{j+1}\mapsto x$ is executed $n-1$ times since the outer loop over $k$ is executed for the $n-1$ values $k=1,2,\dots,n-1$. To work out how often we execute the statement $b_{j+1}\mapsto b_j$, observe that in the worst case we have to loop over $j=k-1,k-2,\dots,0$ for a given value of $k$. Hence we need to execute the statement $b_{j+1}\mapsto b_j$ no more than $k$ times. This allows us to write down a bound on $E(n)$ as follows:

$$
E(n) \le 2(n-1) + \sum_{k=1}^{n-1} k = 2(n-1) + \frac{1}{2}n(n-1) = \frac{1}{2}n^2+\frac{3}{2}n-2.
$$

Using big-O notation we find that $E(n)=\mathcal{O}(n^2)$, i.e. in the worst case the runtime will grow approximately quadratically for large $n$. This is what happens if the list is sorted in *decreasing* order to begin with.

### Best and average case complexity
If, however, the list is already sorted to start with we will never enter the while loop, so the statement $b_{j+1}\mapsto b_j$ is never executed. In this case we have that $E(n)=2(n-1)=\mathcal{O}(n)$, so for large $n$ the runtime will only grow linearly with $n$. Obviously, in this case there is no point in running Insertion Sort in the first place. In general, the runtime will lie somewhere between those two extremes. It is possible to show that the *average* runtime is in fact also quadratic, i.e. $\Theta(n^2)$. In other words if we have a list of numbers which are in random order, then the most likely outcome is that the runtime grows quadratically with the length of this list.

## Measured runtimes
The following code generates random lists with $n=1000,2000,4000$ elements, and then uses InsertionSort to sort these lists. Again we measure the time with the `%timeit` statement (don't forget to execute the cell which defined the `insertion_sort` function first):

```python
import numpy as np
np.random.seed(2141257)

for n in (1000,2000,4000):
    a_list = np.array(range(n))
    np.random.shuffle(a_list)
    %timeit insertion_sort(a_list)
```

The runtime increases by a factor of approximately four when the problem size $n$ doubles. The following plot shows the runtime for different problem sizes:

![Runtime InsertionSort for problems of different size $n$](figures/insertionsort_runtime_linear.png)

We see that the time grows approximately quadratically with $n$, as expected by our theoretical complexity analysis.

This can also be visualised in a slightly different way: if we plot the runtime per element, i.e. $t/n$ then this should be a straight line since $t$ grows quadratically with $n$. The following figure confirms this.

![Average runtime per element of InsertionSort for problems of different size $n$](figures/sorting_runtime_per_element_linear.png)

The quadratic growth in the runtime make Insertion Sort unsuitable for sorting very large lists. As we will see later, it is possible to find a sorting algorithm which only has a $\mathcal{O}(n\log_2(n))$ complexity in both the worst- and average case.

However, Insertion Sort might still be useful if we know that the list that we are sorting is almost in ascending order.
